# 1\. Tuplas

## 1.1 Tupla

Una tupla es una colección ordenada e inmutable de elementos. Se parece a una lista, pero una vez creada no se puede modificar: no se le pueden agregar, quitar ni cambiar elementos.

| Característica | Lista | Tupla |
|----------------|:-----:|:-----:|
| Se escribe con | `[ ]` | `( )` |
| Ordenada (tiene índices) | ✔ | ✔ |
| Permite duplicados | ✔ | ✔ |
| Se puede modificar | ✔ | ✘ |

Se usan para agrupar datos que van juntos y no deberían cambiar, como las coordenadas de un punto, una fecha o los datos fijos de un registro.



## 1.2 Crear Tuplas

* Las tuplas se utilizan para almacenar varios elementos en una sola variable.
* Las listas se crean utilizando parentesis `()`.
* Los elementos de las tuplas están ordenados, no se pueden cambiar y permiten valores duplicados.
* Los elementos de la tupla están indexados, el primer elemento tiene índice [0], el segundo elemento tiene índice [1], etc.

### Ejemplos

```python
coordenada = (6.2442, -75.5812)        # latitud y longitud de Medellín
fecha = (10, "octubre", 2026)          # puede mezclar tipos de datos
vacia = ()
numeros = tuple([5, 1, 6, 8])          # a partir de una lista

print(coordenada)  # (6.2442, -75.5812)
print(numeros)     # (5, 1, 6, 8)
```

Los paréntesis son opcionales: lo que realmente crea la tupla son las **comas**.

```python
punto = 3, 4
print(punto)        # (3, 4)
print(type(punto))  # <class 'tuple'>

## 1.3 Tupla de un solo elemento

Por la regla anterior, unos paréntesis sin coma no crean una tupla: Python los interpreta como paréntesis de una operación matemática.

```python
a = (5)
b = (5,)
print(type(a))  # <class 'int'>    ← es solo el número 5
print(type(b))  # <class 'tuple'>  ← la coma la convierte en tupla
```

## 1.4 Acceder a los elementos

Las tuplas usan índices y rebanado igual que las listas:

```python
dias = ("lunes", "martes", "miércoles", "jueves", "viernes")

print(dias[0])         # lunes
print(dias[-1])        # viernes
print(dias[1:3])       # ('martes', 'miércoles')  ← el rebanado devuelve otra tupla
print(len(dias))       # 5
print("sábado" in dias)  # False
```

## 1.5 Inmutabilidad

Intentar cambiar un elemento produce un error:

```python
dias = ("lunes", "martes", "miércoles")
dias[0] = "domingo"
# TypeError: 'tuple' object does not support item assignment
```

Tampoco existen `append()`, `remove()`, `sort()` ni ningún otro método que modifique la tupla. Si se necesita un cambio, hay que crear una tupla nueva:

```python
dias = ("lunes", "martes", "miércoles")

# Opción 1: concatenar
dias = dias + ("jueves",)
print(dias)  # ('lunes', 'martes', 'miércoles', 'jueves')

# Opción 2: convertir a lista, modificar y volver a tupla
temporal = list(dias)
temporal[0] = "domingo"
dias = tuple(temporal)
print(dias)  # ('domingo', 'martes', 'miércoles', 'jueves')
```

> En la opción 1, `dias = dias + ("jueves",)` no modifica la tupla original: crea una nueva y la asigna a la misma variable.


## 1.6 Métodos de las tuplas

Como no se pueden modificar, las tuplas solo tienen dos métodos:

```python
notas = (4.5, 3.2, 4.5, 2.8, 4.5)

print(notas.count(4.5))  # 3  ← cuántas veces aparece
print(notas.index(3.2))  # 1  ← posición de la primera aparición
```

Las funciones generales sí funcionan, porque no modifican la tupla: `len()`, `sum()`, `max()`, `min()` y `sorted()`. Esta última devuelve una lista:

```python
notas = (4.5, 3.2, 2.8)
print(sum(notas))     # 10.5
print(sorted(notas))  # [2.8, 3.2, 4.5]  ← lista nueva, la tupla no cambia
```

## 1.7 Recorrer una tupla

```python
colores = ("amarillo", "azul", "rojo")
for i, color in enumerate(colores):
    print(i, color)
# 0 amarillo
# 1 azul
# 2 rojo
```

## 1.8 Desempaquetado

Desempaquetar es asignar cada elemento de una tupla a una variable distinta, en una sola línea:

```python
coordenada = (6.2442, -75.5812)
latitud, longitud = coordenada
print(latitud)   # 6.2442
print(longitud)  # -75.5812
```

La cantidad de variables debe coincidir con la cantidad de elementos. Si no, se produce `ValueError`. Con `*` una variable recoge "el resto" de los elementos en una lista:

```python
primero, *resto = (10, 20, 30, 40)
print(primero)  # 10
print(resto)    # [20, 30, 40]

*inicio, ultimo = (10, 20, 30, 40)
print(inicio)   # [10, 20, 30]
print(ultimo)   # 40
```

El desempaquetado es lo que permite intercambiar dos variables en una línea, como en el método de la burbuja:

```python
a = 1
b = 2
a, b = b, a   # a la derecha se crea la tupla (2, 1) y luego se desempaqueta
print(a, b)   # 2 1
```

## 1.9 Lista de tuplas

Una lista de tuplas es una forma común de representar una tabla: la lista puede crecer, pero cada registro queda protegido contra cambios accidentales.

```python
ciudades = [("Medellín", 6.2442, -75.5812),
            ("Bogotá", 4.7110, -74.0721),
            ("Cali", 3.4516, -76.5320)]

for nombre, lat, lon in ciudades:
    print(f"{nombre:<10}{lat:>9.4f}{lon:>10.4f}")
# Medellín     6.2442  -75.5812
# Bogotá       4.7110  -74.0721
# Cali         3.4516  -76.5320

ciudades.append(("Pasto", 1.2136, -77.2811))  # la lista sí se puede modificar
print(len(ciudades))  # 4
```

## 1.10 Una tupla con una lista adentro

La tupla impide cambiar qué objetos contiene, pero si uno de esos objetos es una lista, la lista por dentro sí se puede modificar:

```python
registro = ("Ana", [4.5, 3.8])
registro[1].append(5.0)
print(registro)  # ('Ana', [4.5, 3.8, 5.0])  ← la lista interna cambió
```

## 1.11 ¿Cuándo usar una tupla en lugar de una lista?

- Cuando los datos no deben cambiar (coordenadas, fechas, configuración fija). La tupla protege contra modificaciones por error.
- Cuando cada posición tiene un significado distinto, como `(nombre, edad, nota)`. Las listas se usan más para colecciones de elementos del mismo tipo, como una lista de notas.


# 2\. Conjuntos

## 2.1 Conjunto

Un conjunto (`set`) es una colección sin orden y sin elementos repetidos. Funciona como los conjuntos de las matemáticas: permite calcular uniones, intersecciones y diferencias.

| Característica | Conjunto |
|----------------|:--------:|
| Se escribe con | `{ }` |
| Ordenado (tiene índices) | ✘ |
| Permite duplicados | ✘ |
| Se puede modificar | ✔ |
| Sus elementos deben ser inmutables | ✔ |

## 2.2 Crear conjuntos

Los valores repetidos se eliminan automáticamente

```python
primos = {2, 3, 5, 7}
print(primos)  # {2, 3, 5, 7}

# Los repetidos se eliminan automáticamente
numeros = {5, 1, 6, 8, 1, 3, 5}
print(numeros)  # {1, 3, 5, 6, 8}

# A partir de una lista
lista = [5, 1, 6, 8, 1, 3, 5]
sin_repetidos = set(lista)
print(sin_repetidos)       # {1, 3, 5, 6, 8}
print(len(sin_repetidos))  # 5
```

## 2.3 El conjunto vacío

Las llaves vacías `{}` crean un diccionario vacío, no un conjunto. El conjunto vacío se crea con `set()`:

```python
a = {}
b = set()
print(type(a))  # <class 'dict'>
print(type(b))  # <class 'set'>
```

## 2.4 Los conjuntos no tienen orden

Los elementos no tienen posición, así que no se puede usar índices ni rebanado:

```python
heroes = {"Batman", "Flash", "SuperMan"}
heroes[0]
# TypeError: 'set' object is not subscriptable
```

Además, el orden en que Python muestra los elementos no está garantizado: con textos, puede cambiar de una ejecución a otra. Por eso, cuando se necesita un orden fijo para imprimir, se usa `sorted()`, que devuelve una lista ordenada:

```python
heroes = {"SuperMan", "Batman", "Flash"}
print(sorted(heroes))  # ['Batman', 'Flash', 'SuperMan']
```

## 2.5 Verificar si un elemento pertenece

El operador `in` es la operación más usada con conjuntos, y es mucho más rápida que en una lista. En una lista Python revisa los elementos uno por uno; en un conjunto llega directamente al lugar donde estaría el elemento.

```python
vocales = {"a", "e", "i", "o", "u"}
print("e" in vocales)      # True
print("x" in vocales)      # False
print("x" not in vocales)  # True
```

## 2.6 Agregar y eliminar elementos

```python
numeros = {1, 2, 3}

numeros.add(4)            # agrega un elemento
numeros.add(2)            # ya existe: no pasa nada
print(numeros)            # {1, 2, 3, 4}

numeros.update([5, 6, 1]) # agrega varios elementos de otra colección
print(numeros)            # {1, 2, 3, 4, 5, 6}

numeros.remove(6)         # elimina; si no existe produce KeyError
numeros.discard(99)       # elimina si existe; si no, no hace nada
print(numeros)            # {1, 2, 3, 4, 5}

numeros.clear()           # vacía el conjunto
print(numeros)            # set()
```

> Usa `discard()` cuando no estés seguro de que el elemento exista, y `remove()` cuando su ausencia indique un error que quieras detectar.


## 2.7 Operaciones entre conjuntos

Supongamos dos grupos de estudiantes:

```python
python = {"Ana", "Luis", "Sofía", "Carlos"}
calculo = {"Luis", "Sofía", "Mariana"}
```

| Operación | Operador | Método | Resultado |
|-----------|:--------:|--------|-----------|
| **Unión**: están en alguno de los dos | `a \| b` | `a.union(b)` | todos los estudiantes |
| **Intersección**: están en ambos | `a & b` | `a.intersection(b)` | ven las dos materias |
| **Diferencia**: están en `a` pero no en `b` | `a - b` | `a.difference(b)` | solo ven Python |
| **Diferencia simétrica**: están en uno solo | `a ^ b` | `a.symmetric_difference(b)` | ven una sola materia |

```python
print(sorted(python | calculo))  # ['Ana', 'Carlos', 'Luis', 'Mariana', 'Sofía']
print(sorted(python & calculo))  # ['Luis', 'Sofía']
print(sorted(python - calculo))  # ['Ana', 'Carlos']
print(sorted(calculo - python))  # ['Mariana']
print(sorted(python ^ calculo))  # ['Ana', 'Carlos', 'Mariana']
```

> La diferencia no es simétrica: `python - calculo` y `calculo - python` dan resultados distintos. La unión, la intersección y la diferencia simétrica sí dan lo mismo en cualquier orden.

Las mismas operaciones con números:

```python
a = {1, 2, 3, 4, 5}
b = {4, 5, 6, 7}
print(a | b)  # {1, 2, 3, 4, 5, 6, 7}
print(a & b)  # {4, 5}
print(a - b)  # {1, 2, 3}
print(a ^ b)  # {1, 2, 3, 6, 7}
```

## 2.8 Recorrer un conjunto

```python
vocales = {"u", "a", "o", "e", "i"}
for v in sorted(vocales):
    print(v, end=" ")
print()
# a e i o u
```

## 2.9 Usos típicos

**Eliminar repetidos** de una lista, cuando el orden no importa:

```python
correos = ["ana@unal.edu.co", "luis@unal.edu.co", "ana@unal.edu.co"]
unicos = list(set(correos))
print(len(unicos))  # 2
```

> `set()` no conserva el orden original. Si el orden importa, se usa el método con el ciclo `for` del Ejercicio 10 de la unidad anterior, o `list(dict.fromkeys(lista))` (sección 2.29).

**Detectar repetidos** recordando lo que ya se vio:

```python
numeros = [5, 1, 6, 8, 1, 3, 5]
vistos = set()
repetidos = set()
for n in numeros:
    if n in vistos:
        repetidos.add(n)
    else:
        vistos.add(n)
print(repetidos)  # {1, 5}
```

**Comparar dos colecciones**, sin importar el orden ni los repetidos:

```python
respuesta = ["rojo", "azul", "amarillo"]
correcta = ["amarillo", "azul", "rojo"]
print(respuesta == correcta)            # False ← las listas comparan el orden
print(set(respuesta) == set(correcta))  # True  ← los conjuntos no
```

## 2.10 Los elementos deben ser inmutables

Un conjunto puede contener números, textos y tuplas, pero no listas ni otros conjuntos, porque estos pueden cambiar:

```python
puntos = {(0, 0), (1, 2), (0, 0)}
print(len(puntos))  # 2  ← las tuplas sí se permiten y el repetido se elimina
```

```python
malo = {[1, 2], [3, 4]}
# TypeError: unhashable type: 'list'
```

> "*Unhashable*" significa que Python no puede calcular una huella fija (*hash*) del objeto, porque su contenido podría cambiar. Esa huella es la que permite que `in` sea tan rápido.


# 3\. Diccionarios

## 3.1 Diccionario

Un diccionario (`dict`) es una colección de pares clave: valor. En lugar de acceder a los datos por su posición, como en las listas, se accede por su clave, igual que en un diccionario de papel se busca una palabra para encontrar su significado.

| Característica | Diccionario |
|----------------|:-----------:|
| Se escribe con | `{clave: valor}` |
| Se accede por | la clave |
| Claves repetidas | ✘ (cada clave es única) |
| Valores repetidos | ✔ |
| Se puede modificar | ✔ |
| Las claves deben ser inmutables | ✔ |
| Conserva el orden de inserción | ✔ (desde Python 3.7) |

## 3.2 Crear diccionarios

* Los diccionarios se utilizan para almacenar valores de datos en pares `clave:valor` en una sola variable.
* Los diccionarios se crean utilizando llaves {}.
* Los elementos de los diccionarios están ordenados (a partir de Python 3.7), se pueden cambiar y no permiten claves duplicadas.
* Los elementos del diccionario están referenciados por sus claves, en lugar de un índice numérico (por ejemplo, diccionario["clave"]).

```python
estudiante = {"nombre": "Ana", "edad": 20, "nota": 4.5}
print(estudiante)  # {'nombre': 'Ana', 'edad': 20, 'nota': 4.5}

vacio = {}
precios = dict(cafe=3500, arepa=2000)  # con la función dict()
print(precios)  # {'cafe': 3500, 'arepa': 2000}
```

Paso a paso del primer ejemplo:

1. `"nombre"`, `"edad"` y `"nota"` son las claves.
2. `"Ana"`, `20` y `4.5` son los valores.
3. Cada clave se separa de su valor con `:` y cada par se separa del siguiente con `,`.


## 3.3 Acceder a los valores

```python
estudiante = {"nombre": "Ana", "edad": 20, "nota": 4.5}
print(estudiante["nombre"])  # Ana
print(estudiante["nota"])    # 4.5
```

Si la clave no existe se produce un error:

```python
print(estudiante["ciudad"])
# KeyError: 'ciudad'
```

El método `get()` evita el error: devuelve `None`, o un valor por defecto que uno elige, cuando la clave no existe:

```python
estudiante = {"nombre": "Ana", "edad": 20, "nota": 4.5}
print(estudiante.get("ciudad"))                # None
print(estudiante.get("ciudad", "Sin ciudad"))  # Sin ciudad
print(estudiante.get("nombre", "Sin nombre"))  # Ana  ← la clave existe
```


## 3.4 Agregar y modificar

La misma instrucción sirve para las dos cosas: si la clave no existe, se agrega; si ya existe, se reemplaza su valor.

```python
estudiante = {"nombre": "Ana", "edad": 20}

estudiante["ciudad"] = "Medellín"  # clave nueva: se agrega
estudiante["edad"] = 21            # clave existente: se modifica
print(estudiante)  # {'nombre': 'Ana', 'edad': 21, 'ciudad': 'Medellín'}
```

`update()` agrega o modifica varios pares a la vez:

```python
estudiante.update({"edad": 22, "nota": 4.8})
print(estudiante)  # {'nombre': 'Ana', 'edad': 22, 'ciudad': 'Medellín', 'nota': 4.8}
```

## 3.5 Eliminar

```python
estudiante = {"nombre": "Ana", "edad": 20, "nota": 4.5, "ciudad": "Medellín"}

del estudiante["ciudad"]          # elimina el par; KeyError si no existe
nota = estudiante.pop("nota")     # elimina el par y devuelve su valor
print(nota)                       # 4.5
print(estudiante)                 # {'nombre': 'Ana', 'edad': 20}

print(estudiante.pop("correo", "No existía"))  # No existía ← valor por defecto, sin error

estudiante.clear()
print(estudiante)                 # {}
```

##3.6 Verificar si una clave existe

El operador `in` revisa las claves, no los valores:

```python
estudiante = {"nombre": "Ana", "edad": 20}
print("nombre" in estudiante)           # True
print("Ana" in estudiante)              # False ← "Ana" es un valor, no una clave
print("Ana" in estudiante.values())     # True
```

## 3.7 Recorrer un diccionario

Hay tres métodos que permiten recorrer un diccionario:

| Método | Entrega |
|--------|---------|
| `d.keys()` | las claves |
| `d.values()` | los valores |
| `d.items()` | tuplas `(clave, valor)` |

```python
precios = {"café": 3500, "arepa": 2000, "buñuelo": 1500}

for producto in precios:          # recorrer el diccionario es recorrer sus claves
    print(producto)
# café
# arepa
# buñuelo

for valor in precios.values():
    print(valor)
# 3500
# 2000
# 1500

for producto, precio in precios.items():  # desempaquetado de tuplas (sección 2.27)
    print(f"{producto:<10}${precio:,}")
# café      $3,500
# arepa     $2,000
# buñuelo   $1,500
```

Con las vistas se pueden usar las funciones de siempre:

```python
precios = {"café": 3500, "arepa": 2000, "buñuelo": 1500}
print(len(precios))           # 3
print(sum(precios.values()))  # 7000
print(list(precios.keys()))   # ['café', 'arepa', 'buñuelo']
print(sorted(precios))        # ['arepa', 'buñuelo', 'café']  ← claves ordenadas
```

## 3.8 Usos Comunes de diccionarios


### 3.8.1 Contar secuencias

Es uno de los usos más importantes de los diccionarios. La clave es el elemento y el valor es cuántas veces aparece:

```python
votos = ["azul", "rojo", "azul", "verde", "azul", "rojo"]

conteo = {}
for voto in votos:
    if voto in conteo:
        conteo[voto] += 1   # ya existía: se suma 1
    else:
        conteo[voto] = 1    # primera vez: empieza en 1
print(conteo)  # {'azul': 3, 'rojo': 2, 'verde': 1}
```

Con `get()` el mismo patrón se escribe en una línea:

```python
conteo = {}
for voto in votos:
    conteo[voto] = conteo.get(voto, 0) + 1
print(conteo)  # {'azul': 3, 'rojo': 2, 'verde': 1}
```

Paso a paso con el primer voto, `"azul"`:

1. `conteo.get("azul", 0)` devuelve `0`, porque la clave todavía no existe.
2. Se le suma 1 y se guarda: `conteo["azul"] = 1`.
3. La próxima vez que aparezca `"azul"`, `get()` devolverá `1` y se guardará `2`.

> `collections.Counter` hace exactamente esto. De hecho, un `Counter` es un tipo especial de diccionario.


### 3.8.2 Agrupar

Aquí el valor de cada clave es una lista con los elementos que pertenecen a ese grupo:

```python
heroes = ["Batman", "SuperMan", "Flash", "Shazam", "Aquaman", "Batgirl"]

por_letra = {}
for heroe in heroes:
    letra = heroe[0]
    if letra not in por_letra:
        por_letra[letra] = []      # primera vez: se crea la lista vacía
    por_letra[letra].append(heroe)
print(por_letra)
# {'B': ['Batman', 'Batgirl'], 'S': ['SuperMan', 'Shazam'], 'F': ['Flash'], 'A': ['Aquaman']}
```

### 3.8.3 Tablas con nombre

En la clase 05 las tablas eran listas de listas y había que recordar que la posición para obtener un valor. Con una lista de diccionarios, cada campo se accede por su nombre:

```python
estudiantes = [{"nombre": "Ana", "edad": 20, "nota": 4.5},
               {"nombre": "Luis", "edad": 22, "nota": 3.2},
               {"nombre": "Sofía", "edad": 19, "nota": 4.8}]

for e in estudiantes:
    print(f"{e['nombre']:<8}{e['nota']:>5}")
# Ana       4.5
# Luis      3.2
# Sofía     4.8

aprobados = [e["nombre"] for e in estudiantes if e["nota"] >= 4.0]
print(aprobados)  # ['Ana', 'Sofía']

mejor = sorted(estudiantes, key=lambda e: e["nota"], reverse=True)[0]
print(mejor["nombre"])  # Sofía
```

> Dentro de un f-string con comillas dobles, las claves se escriben con comillas simples: `f"{e['nombre']}"`.


### 3.8.4 Diccionario de diccionarios

Cuando cada registro tiene un identificador único, como un código, se puede usar como clave para encontrarlo directamente, sin recorrer la tabla:

```python
inventario = {101: {"nombre": "Arroz", "cantidad": 20, "precio": 3500},
              102: {"nombre": "Aceite", "cantidad": 3, "precio": 12000}}

print(inventario[102]["nombre"])  # Aceite

inventario[102]["cantidad"] += 10
print(inventario[102])  # {'nombre': 'Aceite', 'cantidad': 13, 'precio': 12000}

for codigo, datos in inventario.items():
    print(codigo, datos["nombre"], datos["cantidad"])
# 101 Arroz 20
# 102 Aceite 13
```

cuadrados = {n: n ** 2 for n in range(1, 6)}

productos = ["café", "arepa", "buñuelo"]

catalogo = {p: c for p, c in zip(productos, precios)}

baratos = {p: c for p, c in catalogo.items() if c < 3000}

# Aplicar un aumento del 10 %
con_aumento = {p: round(c * 1.1) for p, c in catalogo.items()}
print(con_aumento)  # {'café': 3850, 'arepa': 2200, 'buñuelo': 1650}
```



## 3.10 Claves válidas

Las claves deben ser inmutables, igual que los elementos de un conjunto: números, textos y tuplas sirven; las listas no.

```python
distancias = {("Medellín", "Bogotá"): 415,
              ("Medellín", "Cali"): 420}
print(distancias[("Medellín", "Bogotá")])  # 415
```

```python
malo = {[1, 2]: "lista"}
# TypeError: unhashable type: 'list'
```

Los valores, en cambio, pueden ser de cualquier tipo: listas, conjuntos u otros diccionarios.


# 4\. Resumen de estructuras

| | Lista | Tupla | Conjunto | Diccionario |
|---|:---:|:---:|:---:|:---:|
| Sintaxis | `[1, 2]` | `(1, 2)` | `{1, 2}` | `{"a": 1}` |
| Vacío | `[]` | `()` | `set()` | `{}` |
| Ordenado | ✔ | ✔ | ✘ | ✔ (inserción) |
| Modificable | ✔ | ✘ | ✔ | ✔ |
| Duplicados | ✔ | ✔ | ✘ | claves ✘, valores ✔ |
| Se accede por | índice | índice | — (solo `in`) | clave |

- **Lista**: una secuencia que va a cambiar y donde importa el orden.
- **Tupla**: un grupo fijo de datos relacionados, como un registro o unas coordenadas.
- **Conjunto**: cuando importa *si* algo está y no cuántas veces ni en qué orden; operaciones de unión, intersección y diferencia.
- **Diccionario**: cuando se quiere buscar información por un nombre o identificador, contar o agrupar.
