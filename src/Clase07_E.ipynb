# 1\. Strings

Un string o cadena de caracteres es el tipo de dato que Python usa para representar texto: un nombre, una frase, una dirección de correo, el contenido de un archivo. Su tipo se llama `str`, abreviatura de *string*, que en inglés significa "cadena" o "hilera", porque los caracteres van encadenados uno tras otro.

Formalmente, un string es una secuencia ordenada e inmutable de caracteres. Cada parte de esa definición es importante:

- Secuencia: los caracteres están uno detrás de otro, como las cuentas de un collar. Python puede recorrerlos, contarlos y acceder a cualquiera de ellos.
- Ordenada: cada carácter ocupa una posición fija, llamada índice, y el orden importa. `"amor"` y `"roma"` tienen las mismas letras, pero son strings distintos.
- Inmutable: una vez creado, el string no se puede modificar. Las operaciones que parecen cambiarlo en realidad crean un string nuevo.

Un carácter es la unidad mínima de texto: una letra, un dígito, un espacio, un signo de puntuación o un carácter especial como el salto de línea. Python no tiene un tipo de dato separado para un carácter; un carácter es simplemente un string de longitud 1.

Internamente, el computador no guarda letras sino números. Python identifica cada carácter con un número según el estándar Unicode, que asigna un código único a los caracteres de casi todos los idiomas del mundo. Gracias a esto, un string puede contener sin problema la ñ, las vocales con tilde, emojis o letras de otros alfabetos. Las funciones `ord()` y `chr()` permiten pasar de un carácter a su código y viceversa:

```python
print(type("Hola"))  # <class 'str'>
print(type("a"))     # <class 'str'>  ← un carácter también es un str
print(ord("A"))      # 65  ← código Unicode de la letra A
print(ord("a"))      # 97
print(chr(241))      # ñ   ← carácter que corresponde al código 241
```

> Estos códigos explican por qué, al ordenar textos, las mayúsculas quedan antes que las minúsculas: el código de `"A"` (65) es menor que el de `"a"` (97).


## 1.1 Relación entre strings y tuplas

En Python, las listas, las tuplas y los strings pertenecen a una misma familia de tipos llamada secuencias: colecciones de elementos ordenados a los que se accede por su posición. Dentro de esa familia, el string es el pariente más cercano de la tupla, porque ambos son inmutables: ni a una tupla ni a un string se les puede cambiar, agregar o quitar un elemento después de creados.

Por eso se puede pensar en un string como una tupla especializada, cuyos elementos son siempre caracteres y que además trae herramientas propias para trabajar con texto. Esta relación tiene una consecuencia práctica muy útil: casi todo lo que ya sabes hacer con tuplas funciona exactamente igual con strings:

| Operación | Tupla | String |
|-----------|-------|--------|
| Crear | `t = ("h", "o", "l", "a")` | `s = "hola"` |
| Longitud | `len(t)` → `4` | `len(s)` → `4` |
| Índice | `t[0]` → `'h'` | `s[0]` → `'h'` |
| Rebanado | `t[1:3]` → `('o', 'l')` | `s[1:3]` → `'ol'` |
| Pertenencia | `"o" in t` → `True` | `"o" in s` → `True` |
| Recorrer | `for x in t:` | `for c in s:` |
| Concatenar | `t + ("s",)` | `s + "s"` |
| Repetir | `t * 2` | `s * 2` |
| Contar | `t.count("a")` | `s.count("a")` |
| Buscar posición | `t.index("l")` | `s.index("l")` |
| Modificar un elemento | ✘ `TypeError` | ✘ `TypeError` |

Las diferencias principales son dos:

- Los elementos de una tupla pueden ser de cualquier tipo; los de un string son siempre caracteres.
- Los strings tienen muchos métodos propios para trabajar con texto (`upper()`, `split()`, `find()`, `strip()`...), que se explicará más adelante.

Se puede pasar de uno al otro:

```python
t = tuple("hola")
print(t)  # ('h', 'o', 'l', 'a')

s = "".join(t)
print(s)  # hola

a, b, c = "xyz"   # un string también se puede desempaquetar
print(a, b, c)    # x y z
```

````python
t = tuple("hola")
print(t)  # ('h', 'o', 'l', 'a')

s = "".join(t)
print(s)  # hola

a, b, c = "xyz"   # un string también se puede desempaquetar
print(a, b, c)    # x y z
````

## 1.2 Crear strings

Un string se crea escribiendo el texto entre comillas. A un texto escrito directamente en el código de esta forma se le llama literal de string. Python acepta comillas simples (`'...'`) o dobles (`"..."`) indistintamente; lo único obligatorio es que el texto empiece y termine con el mismo tipo de comilla. Las comillas no forman parte del string: solo le indican a Python dónde empieza y dónde termina el texto.

Tener dos tipos de comillas permite escribir una de ellas dentro del texto sin confundir a Python. También se puede crear un string a partir de otro tipo de dato con la función `str()`. El string vacío, `""`, no contiene ningún carácter y tiene longitud 0; es el punto de partida habitual cuando se construye un texto agregando caracteres uno por uno.

```python
nombre = "Ana"                 # comillas dobles
ciudad = 'Medellín'            # comillas simples: es exactamente lo mismo
frase = "It's a test"          # comillas dobles para poder usar ' adentro
cita = 'Ella dijo "hola"'      # y al revés
numero = str(2026)             # convertir otro tipo a string
vacio = ""

print(numero, type(numero))  # 2026 <class 'str'>
print(len(vacio))            # 0
```

Las comillas triples (`"""..."""` o `'''...'''`) permiten escribir un texto que ocupa varias líneas. Los saltos de línea que se escriben dentro de ellas quedan guardados como parte del string:

```python
poema = """Caminante, son tus huellas
el camino y nada más."""
print(poema)
# Caminante, son tus huellas
# el camino y nada más.

````python
nombre = "Ana"                 # comillas dobles
ciudad = 'Medellín'            # comillas simples: es exactamente lo mismo
frase = "It's a test"          # comillas dobles para poder usar ' adentro
cita = 'Ella dijo "hola"'      # y al revés
numero = str(2026)             # convertir otro tipo a string
vacio = ""

print(numero, type(numero))  # 2026 <class 'str'>
print(len(vacio))            # 0

poema = """Caminante, son tus huellas
el camino y nada más."""
print(poema)
# Caminante, son tus huellas
# el camino y nada más.
````

## 1.3 Caracteres especiales (secuencias de escape)

Algunos caracteres no se pueden escribir directamente dentro de un string, porque son invisibles (como un salto de línea o una tabulación) o porque tienen un significado especial para Python (como las comillas que delimitan el texto). Para escribirlos se usan las secuencias de escape: combinaciones de caracteres que empiezan con una barra invertida `\`.

La barra invertida le indica a Python que el carácter que viene a continuación no se debe tomar literalmente, sino con un significado especial. Aunque una secuencia de escape se escribe con dos símbolos, representa un solo carácter dentro del string.

| Secuencia | Significado |
|-----------|-------------|
| `\n` | Salto de línea |
| `\t` | Tabulación |
| `\\` | Una barra invertida |
| `\"` y `\'` | Comillas dentro de un texto con las mismas comillas |

```python
print("Línea 1\nLínea 2")
# Línea 1
# Línea 2

print("Nombre:\tAna")
# Nombre:	Ana

print("C:\\Usuarios\\ana")
# C:\Usuarios\ana

print(len("\n"))  # 1  ← aunque se escribe con dos símbolos, es un solo carácter
```

> El salto de línea `\n` es muy importante, es el carácter que separa las líneas de un archivo de texto.


````python
print("Línea 1\nLínea 2")
# Línea 1
# Línea 2

print("Nombre:\tAna")
# Nombre:   Ana

print("C:\\Usuarios\\ana")
# C:\Usuarios\ana

print(len("\n"))  # 1  ← aunque se escribe con dos símbolos, es un solo carácter
````

## 1.4 Longitud: `len()`

La **longitud** de un string es la cantidad de caracteres que contiene. Se obtiene con la función `len()`, la misma que se usa con listas y tuplas. Todos los caracteres cuentan: las letras, los dígitos, los espacios, los signos de puntuación y los caracteres especiales como `\n`. La longitud del string vacío es 0.

La longitud es importante porque define el rango de **índices válidos**: en un string de longitud `n`, las posiciones van de `0` a `n - 1`.

```python
texto = "Programación"
print(len(texto))          # 12
print(len("Hola mundo"))   # 10  ← el espacio también es un carácter
```

````python
texto = "Programación"
print(len(texto))          # 12
print(len("Hola mundo"))   # 10  ← el espacio también es un carácter
````

## 1.5 Índices

Un índice es el número que identifica la posición de un carácter dentro del string. Para obtener un carácter se escribe el nombre del string seguido del índice entre corchetes: `texto[i]`. El resultado es siempre un string de un solo carácter. Esta forma se conoce como Acceso Indexado.

Python numera las posiciones empezando en 0: el primer carácter está en la posición 0, el segundo en la posición 1, y el último en la posición `len(texto) - 1`. También acepta índices negativos, que cuentan desde el final: `-1` es el último carácter, `-2` el penúltimo, y así sucesivamente. Los índices negativos son útiles para llegar al final del texto sin necesidad de conocer su longitud.

| Carácter | P | y | t | h | o | n |
|----------|:-:|:-:|:-:|:-:|:-:|:-:|
| Índice positivo | 0 | 1 | 2 | 3 | 4 | 5 |
| Índice negativo | -6 | -5 | -4 | -3 | -2 | -1 |

```python
lenguaje = "Python"
print(lenguaje[0])    # P
print(lenguaje[3])    # h
print(lenguaje[-1])   # n  ← el último carácter
print(lenguaje[len(lenguaje) - 1])  # n  ← equivalente, pero más largo
```

Un índice fuera del rango produce un error:

```python
lenguaje = "Python"
print(lenguaje[6])
# IndexError: string index out of range
```


````python
lenguaje = "Python"
print(lenguaje[0])    # P
print(lenguaje[3])    # h
print(lenguaje[-1])   # n  ← el último carácter
print(lenguaje[len(lenguaje) - 1])  # n  ← equivalente, pero más largo
````

## 1.6 Rebanado (*slicing*)

El rebanado (*slicing*) es la operación que extrae una porción consecutiva de un string. A esa porción se le llama substring o subcadena. Se escribe `texto[inicio:fin:paso]`, donde:

- `inicio` es la posición donde empieza la porción, y se incluye. Si se omite, se empieza desde el principio del string.
- `fin` es la posición donde termina, y se excluye: el rebanado se detiene justo antes de ella. Si se omite, se llega hasta el final.
- `paso` indica de cuántas en cuántas posiciones se avanza. Si se omite, vale 1. Si es negativo, el recorrido se hace de derecha a izquierda.

El resultado es siempre un string nuevo; el original no cambia. El rebanado funciona exactamente igual que en las listas y las tuplas. Una forma de recordar que `fin` se excluye es que, con paso 1, la longitud del resultado es `fin - inicio`: `texto[2:5]` tiene 3 caracteres.

```python
lenguaje = "Python"
print(lenguaje[0:2])   # Py
print(lenguaje[2:])    # thon    ← desde la posición 2 hasta el final
print(lenguaje[:4])    # Pyth    ← desde el inicio hasta la posición 3
print(lenguaje[-3:])   # hon     ← los últimos 3 caracteres
print(lenguaje[::2])   # Pto     ← uno sí, uno no
print(lenguaje[::-1])  # nohtyP  ← el string al revés
```

Paso a paso de `lenguaje[1:5:2]`:

1. Empieza en la posición 1: `"y"`.
2. Avanza de 2 en 2: posición 3, `"h"`.
3. La siguiente sería la posición 5, pero el rebanado se detiene antes de `fin`.
4. Resultado: `"yh"`.

> A diferencia de los índices, el rebanado nunca produce error por salirse del rango: `"Python"[2:100]` devuelve `"thon"` y `"Python"[10:]` devuelve `""`.

Un uso muy común es separar partes de un texto con formato fijo:

```python
fecha = "2026-10-18"
anio = fecha[:4]
mes = fecha[5:7]
dia = fecha[8:]
print(dia, mes, anio)  # 18 10 2026
```

## 1.7 Inmutabilidad

Que un string sea inmutable significa que, una vez creado, su contenido no se puede alterar: no se puede reemplazar, agregar ni eliminar un carácter en su lugar. Esto es igual que en las tuplas y diferente de las listas. Intentar asignar un valor a una posición produce un error:

```python
palabra = "gato"
palabra[0] = "p"
# TypeError: 'str' object does not support item assignment
```

Para "modificarlo" se crea un string nuevo:

```python
palabra = "gato"
palabra = "p" + palabra[1:]
print(palabra)  # pato
```

Aquí es importante distinguir entre el string y la variable. La variable `palabra` es solo un nombre que apunta a un valor. En la segunda línea no se modifica el string `"gato"`: se construye un string nuevo, `"pato"`, uniendo `"p"` con el rebanado `"ato"`, y luego la variable pasa a apuntar a ese string nuevo. El string `"gato"` simplemente deja de usarse.

> Por esta razón, ningún método de los strings modifica el string original: todos devuelven un string nuevo. Esto se repite a lo largo de la sección porque es la fuente del error más común con strings.


## 1.8 Concatenar y repetir

Concatenar significa unir dos o más strings, uno a continuación del otro, para formar un string nuevo. Se hace con el operador `+`. Repetir un string se hace con el operador `*` seguido de un número entero, y produce un string nuevo con el texto original repetido esa cantidad de veces.

Ninguno de los dos operadores agrega espacios automáticamente: si se necesitan, hay que incluirlos dentro del texto.

```python
nombre = "Ana"
saludo = "Hola, " + nombre + "!"
print(saludo)       # Hola, Ana!

print("-" * 20)     # --------------------
print("ja" * 3)     # jajaja
```

Solo se pueden concatenar strings con strings:

```python
edad = 20
print("Tengo " + edad + " años")
# TypeError: can only concatenate str (not "int") to str
```

Se corrige convirtiendo el número con `str()`, o mejor, con un f-string:

```python
edad = 20
print("Tengo " + str(edad) + " años")  # Tengo 20 años
print(f"Tengo {edad} años")            # Tengo 20 años
```

## 1.9 Recorrer un string (*looping*)

Recorrer un string (en inglés, *looping* o *iterar*) significa visitar sus caracteres uno por uno, en orden, para hacer algo con cada uno: imprimirlo, contarlo, compararlo o usarlo para construir otro texto. Como el string es una secuencia, el ciclo `for` lo recorre directamente: en cada vuelta, la variable del ciclo toma el valor del siguiente carácter, desde el primero hasta el último.

```python
for letra in "Hola":
    print(letra)
# H
# o
# l
# a
```

Con `enumerate()` se obtiene también la posición:

```python
for i, letra in enumerate("sol"):
    print(i, letra)
# 0 s
# 1 o
# 2 l
```

Un recorrido permite construir un string nuevo agregando caracteres uno a uno. Por ejemplo, para quitar las vocales:

```python
frase = "Fundamentos de programación"
sin_vocales = ""
for c in frase:
    if c.lower() not in "aeiouáéíóú":
        sin_vocales += c
print(sin_vocales)  # Fndmnts d prgrmcn
```

## 1.10 Contar (*counting*)

Contar consiste en determinar cuántas veces aparece algo en un texto: cuántas vocales tiene una frase, cuántas veces se repite una palabra o cuántos dígitos hay en un código. Hay dos formas de hacerlo.

La primera es el patrón de contador, el mismo que se usa con listas: una variable que empieza en 0 y aumenta en 1 cada vez que un carácter cumple una condición. Es la forma más flexible, porque la condición puede ser cualquiera:

```python
frase = "La programación es divertida"
vocales = 0
for c in frase.lower():
    if c in "aeiouáéíóú":
        vocales += 1
print("Vocales:", vocales)  # Vocales: 11
```

La segunda forma es el método `count()`, que cuenta cuántas veces aparece un substring (un pedazo de texto de uno o varios caracteres) dentro del string. Es más corto, pero solo sirve para buscar un texto exacto, no una condición:

```python
print("banana".count("a"))    # 3
print("banana".count("na"))   # 2
print("banana".count("ana"))  # 1  ← no cuenta coincidencias que se solapan
print("banana".count("x"))    # 0
```

> En `"banana"`, `"ana"` aparece en las posiciones 1 y 3, pero las dos coincidencias comparten la `"a"` de la posición 3. `count()` sigue buscando después de cada coincidencia encontrada, así que solo cuenta una.

`count()` acepta posiciones de inicio y fin, como el rebanado:

```python
print("banana".count("a", 2))     # 2  ← desde la posición 2
print("banana".count("a", 0, 3))  # 1  ← entre las posiciones 0 y 2
```

## 1.11 El operador `in`

El operador `in` es un operador de pertenencia: responde con `True` o `False` a la pregunta "¿está este texto dentro de este otro?". Su negación es `not in`.

Con strings, `in` se comporta un poco distinto que con listas. En una lista, `in` busca un elemento completo. En un string busca un substring, es decir, cualquier secuencia consecutiva de caracteres, aunque no sea una palabra completa. La búsqueda distingue mayúsculas de minúsculas.

```python
frase = "Fundamentos de programación"
print("de" in frase)        # True
print("program" in frase)   # True  ← no tiene que ser una palabra completa
print("Programación" in frase)  # False ← distingue mayúsculas
print("python" not in frase)    # True
```

Un caso típico es revisar si un carácter pertenece a un grupo:

```python
c = "e"
print(c in "aeiou")  # True
```

## 1.12 Comparar strings

Comparar strings significa determinar si son iguales o cuál va primero en un orden. El operador `==` devuelve `True` solo si los dos strings tienen exactamente los mismos caracteres en el mismo orden, incluidas las mayúsculas y los espacios.

Los operadores `<`, `>`, `<=` y `>=` comparan en orden lexicográfico, que es el orden de los diccionarios: se comparan los primeros caracteres de ambos strings; si son iguales, se comparan los segundos, y así sucesivamente hasta encontrar una diferencia. Para decidir qué carácter es "menor", Python usa su código Unicode (el valor que devuelve `ord()`). Es la misma regla que se usa para comparar tuplas elemento por elemento:

```python
print("hola" == "Hola")   # False ← distingue mayúsculas
print("ana" < "luis")     # True  ← 'a' va antes que 'l'
print("casa" < "casas")   # True  ← un texto que es el inicio del otro va primero
print("Zeta" < "alfa")    # True  ← las mayúsculas van antes que las minúsculas
print(("a", "b") < ("a", "c"), "ab" < "ac")  # True True ← misma regla para tuplas y strings
```

> Esta es la misma regla que usa `sort()` para ordenar listas de strings.

## 1.13 ¿Qué es un método?

Un método es una función que pertenece a un tipo de dato y que se aplica sobre un valor concreto de ese tipo. Se llama escribiendo el valor (o la variable que lo contiene), un punto, el nombre del método y paréntesis: `texto.upper()`. Si el método necesita información adicional, se le pasa como argumento dentro de los paréntesis, como en `texto.count("a")`.

La diferencia con una función como `len(texto)` está en la forma de llamarla: la función recibe el string como argumento, mientras que el método se escribe "pegado" al string con un punto. Los strings tienen más de 40 métodos; en esta sección se ven los más usados.

Como los strings son inmutables, ningún método modifica el string sobre el que se aplica. Los métodos que transforman el texto devuelven un string nuevo, y los que hacen preguntas sobre el texto devuelven un número o un valor `True`/`False`.


## 1.14 Métodos para cambiar mayúsculas y minúsculas

Estos métodos devuelven una copia del string con las letras convertidas a mayúsculas o minúsculas según distintas reglas. Solo afectan a las letras: los dígitos, los espacios y los signos quedan iguales.

- `upper()` convierte todas las letras a mayúsculas.
- `lower()` convierte todas las letras a minúsculas.
- `capitalize()` convierte el primer carácter del string a mayúscula y todos los demás a minúscula. Es útil para dar formato a una oración.
- `title()` convierte a mayúscula la primera letra de cada palabra y el resto a minúscula. Es útil para nombres propios y títulos.
- `swapcase()` invierte cada letra: las mayúsculas pasan a minúsculas y viceversa.

| Método | Resultado con `"hOLA mUNDO"` |
|--------|------------------------------|
| `upper()` | `'HOLA MUNDO'` |
| `lower()` | `'hola mundo'` |
| `capitalize()` | `'Hola mundo'` ← primera letra en mayúscula y el resto en minúscula |
| `title()` | `'Hola Mundo'` ← primera letra de cada palabra |
| `swapcase()` | `'Hola Mundo'` ← invierte cada letra |

```python
texto = "hOLA mUNDO"
print(texto.upper())       # HOLA MUNDO
print(texto.lower())       # hola mundo
print(texto.capitalize())  # Hola mundo
print(texto.title())       # Hola Mundo
print(texto.swapcase())    # Hola Mundo
print(texto)               # hOLA mUNDO  ← el original no cambió
```

Funcionan también con letras del español:

```python
print("ñandú".upper())  # ÑANDÚ
```

El uso más común de `lower()` es comparar textos sin distinguir mayúsculas:

```python
respuesta = "SÍ"
if respuesta.lower() == "sí":
    print("Respuesta afirmativa")  # Respuesta afirmativa
```

## 1.15 Error común: los métodos no modifican el string

⚠️

```python
nombre = "ana"
nombre.upper()        # crea "ANA", pero no se guarda en ningún lado
print(nombre)         # ana

nombre = nombre.upper()  # hay que asignar el resultado
print(nombre)            # ANA
```

## 1.16 Métodos para buscar: `find()`, `rfind()` e `index()`

Los métodos de búsqueda responden a la pregunta "¿dónde está este texto?", a diferencia de `in`, que solo responde si está o no.

- `find(sub)` busca el substring `sub` de izquierda a derecha y devuelve el índice donde empieza su primera aparición. Si no lo encuentra, devuelve `-1`. Se eligió `-1` porque las posiciones encontradas siempre son 0 o mayores, así que no se puede confundir con un resultado válido.
- `find(sub, inicio)` empieza a buscar desde la posición `inicio`. Esto permite encontrar las apariciones siguientes a la primera.
- `rfind(sub)` (*right find*) busca de derecha a izquierda y devuelve la posición de la última aparición.
- `index(sub)` funciona igual que `find()`, pero si no encuentra el substring produce un error `ValueError` en lugar de devolver `-1`.

```python
texto = "banana"
print(texto.find("na"))     # 2
print(texto.find("na", 3))  # 4   ← busca a partir de la posición 3
print(texto.rfind("na"))    # 4   ← busca desde el final (la última aparición)
print(texto.find("x"))      # -1  ← no está
```

Ejemplo del error que produce `index()`:

```python
print("banana".index("x"))
# ValueError: substring not found
```

`find()` se combina con el rebanado para extraer partes de un texto:

```python
correo = "peter.pan@unal.edu.co"
arroba = correo.find("@")
usuario = correo[:arroba]
dominio = correo[arroba + 1:]
print(usuario)  # pater.pan
print(dominio)  # unal.edu.co
```

Paso a paso:

1. `correo.find("@")` devuelve `9`, la posición de la arroba.
2. `correo[:9]` toma todo lo que está antes de la arroba.
3. `correo[10:]` toma todo lo que está después de la arroba.

Para encontrar todas las apariciones, se repite `find()` empezando después de la última encontrada:

```python
texto = "la casa de la abuela es la más bonita"
posiciones = []
pos = texto.find("la")
while pos != -1:
    posiciones.append(pos)
    pos = texto.find("la", pos + 1)
print(posiciones)  # [0, 11, 18, 24]
```

## 1.17 Error común: usar `find()` directamente en un `if`

⚠️

```python
texto = "Hola mundo"
if texto.find("Hola"):
    print("Encontrado")
else:
    print("No encontrado")
# No encontrado  ← ¡pero sí está!
```

`find()` devolvió `0` (la posición), y `0` se considera falso en un `if`. Además, `-1` se considera verdadero. Para saber si un texto contiene otro se usa `in`; `find()` se usa cuando se necesita la posición:

```python
texto = "Hola mundo"
if "Hola" in texto:
    print("Encontrado")  # Encontrado

if texto.find("Hola") != -1:   # forma correcta con find()
    print("Encontrado")  # Encontrado
```

## 1.18 Métodos para limpiar: `strip()`, `lstrip()` y `rstrip()`

Limpiar un string significa quitarle los caracteres sobrantes de sus extremos. Los datos reales casi nunca llegan limpios: el usuario escribe espacios de más antes o después de su nombre, y cada línea que se lee de un archivo termina con un `\n`. A estos caracteres invisibles (el espacio, la tabulación `\t` y el salto de línea `\n`) se les llama espacios en blanco.

- `strip()` devuelve una copia del string sin los espacios en blanco del inicio y del final.
- `lstrip()` (*left strip*) los quita solo del inicio (la izquierda).
- `rstrip()` (*right strip*) los quita solo del final (la derecha).

Los tres trabajan únicamente sobre los extremos: nunca tocan los caracteres del medio del texto.

```python
dato = "   Peter Pan  "
print(dato.strip())   # Peter Pan
print(dato.lstrip())  # Peter Pan
print(dato.rstrip())  #    Peter Pan     ← solo la derecha
```


## 1.19 Otros métodos útiles

`replace(viejo, nuevo)` devuelve una copia del string en la que todas las apariciones del substring `viejo` se reemplazan por `nuevo`. Un tercer argumento opcional limita la cantidad de reemplazos:

```python
frase = "Me gusta el café. El café de Colombia."
print(frase.replace("café", "té"))     # Me gusta el té. El té de Colombia.
print(frase.replace("café", "té", 1))  # Me gusta el té. El café de Colombia.  ← solo la primera
print(frase.replace(" ", ""))          # Megustaelcafé.ElcafédeColombia.
```

`startswith(sub)` y `endswith(sub)` responden `True` si el string empieza o termina, respectivamente, con el substring indicado. Son útiles, por ejemplo, para revisar la extensión de un archivo:

```python
archivo = "informe_final.pdf"
print(archivo.endswith(".pdf"))        # True
print(archivo.startswith("informe"))   # True
print(archivo.endswith((".jpg", ".png")))  # False ← acepta una tupla de opciones
```

Los métodos que empiezan por `is` son métodos de validación: revisan qué tipo de caracteres contiene el string y devuelven `True` o `False`. Responden preguntas como "¿este texto es un número?" o "¿solo tiene letras?". Todos devuelven `False` si el string está vacío:

| Método | Devuelve `True` si... | `"abc"` | `"123"` | `"abc123"` | `"  "` |
|--------|-----------------------|:-------:|:-------:|:----------:|:------:|
| `isalpha()` | todos son letras | ✔ | ✘ | ✘ | ✘ |
| `isdigit()` | todos son dígitos | ✘ | ✔ | ✘ | ✘ |
| `isalnum()` | todos son letras o dígitos | ✔ | ✔ | ✔ | ✘ |
| `isspace()` | todos son espacios | ✘ | ✘ | ✘ | ✔ |
| `isupper()` / `islower()` | todas las letras están en mayúscula / minúscula | ✘ / ✔ | ✘ / ✘ | ✘ / ✔ | ✘ / ✘ |

Son muy útiles para validar lo que escribe el usuario antes de convertirlo:

```python
edad = "25"
if edad.isdigit():
    edad = int(edad)
    print("Edad válida:", edad)  # Edad válida: 25
```

> `isdigit()` devuelve `False` para números negativos o decimales, como `"-5"` o `"3.5"`, porque el `-` y el `.` no son dígitos.


## 1.20 Convertir un string en lista: `list()`

La función `list()` convierte cualquier secuencia en una lista. Aplicada a un string, produce una lista donde cada carácter es un elemento, incluidos los espacios y los signos:

```python
letras = list("hola")
print(letras)  # ['h', 'o', 'l', 'a']
```

Esta conversión es la forma de obtener una versión modificable del texto: como las listas no son inmutables, se pueden cambiar, agregar o eliminar caracteres, y al final volver a unir la lista en un string con `join()`:

```python
letras = list("gato")
letras[0] = "p"
letras.append("s")
palabra = "".join(letras)
print(palabra)  # patos
```

## 1.21 Separar un string en palabras: `split()`

El método `split()` (en inglés, "dividir") corta un string en pedazos y devuelve una lista de strings. El texto por donde se corta se llama separador o delimitador, y no se incluye en el resultado.

- Sin argumentos, `split()` usa como separador cualquier secuencia de espacios en blanco (uno o varios espacios, tabulaciones o saltos de línea) e ignora los que haya al inicio o al final. Es la forma más práctica de separar un texto en palabras.
- Con un argumento, como `split(",")`, corta exactamente en cada aparición de ese texto. Es la forma de separar datos que vienen en un formato definido, como una fecha `18/10/2026` o una línea de un archivo CSV.

```python
frase = "  La   programación es\tdivertida \n"
palabras = frase.split()
print(palabras)       # ['La', 'programación', 'es', 'divertida']
print(len(palabras))  # 4  ← cantidad de palabras
```

Con un argumento, separa exactamente en ese texto:

```python
fecha = "18/10/2026"
print(fecha.split("/"))  # ['18', '10', '2026']

registro = "Ana;20;4.5"
nombre, edad, nota = registro.split(";")  # desempaquetado
print(nombre, int(edad), float(nota))     # Ana 20 4.5
```

> `split()` siempre devuelve strings. En el ejemplo, `edad` y `nota` hay que convertirlos con `int()` y `float()` para hacer cálculos.

Diferencia importante entre `split()` y `split(" ")`:

```python
texto = "hola  mundo"
print(texto.split())     # ['hola', 'mundo']
print(texto.split(" "))  # ['hola', '', 'mundo']  ← el doble espacio produce un elemento vacío
print("a,b,,c".split(","))  # ['a', 'b', '', 'c']
```

Con un segundo argumento se limita la cantidad de cortes:

```python
linea = "nombre: Ana: Medellín"
print(linea.split(":", 1))  # ['nombre', ' Ana: Medellín']  ← solo un corte
```

## 1.22 Unir una lista en un string: `join()`

El método `join()` (en inglés, "unir") es la operación contraria a `split()`: recibe una lista (o cualquier secuencia) de strings y los une en un solo string, colocando un separador entre cada par de elementos.

Su forma de escribirse puede parecer al revés al principio: el método se llama sobre el separador, y la lista va entre los paréntesis: `separador.join(lista)`. Por ejemplo, `" ".join(palabras)` se lee como "une las palabras poniendo un espacio entre ellas". El separador solo se coloca entre los elementos, nunca al inicio ni al final.

```python
palabras = ["La", "programación", "es", "divertida"]
print(" ".join(palabras))   # La programación es divertida
print("-".join(palabras))   # La-programación-es-divertida
print("".join(palabras))    # Laprogramaciónesdivertida
print(", ".join(["Ana", "Luis", "Sofía"]))  # Ana, Luis, Sofía
```

Todos los elementos de la lista deben ser strings:

```python
print("-".join([2026, 10, 18]))
# TypeError: sequence item 0: expected str instance, int found
```

```python
numeros = [2026, 10, 18]
print("-".join([str(n) for n in numeros]))  # 2026-10-18
```

Combinando `split()`, una list comprehension y `join()` se transforman textos palabra por palabra:

```python
frase = "el que madruga dios lo ayuda"
invertidas = " ".join([p[::-1] for p in frase.split()])
print(invertidas)  # le euq agurdam soid ol aduya

iniciales = "".join([p[0].upper() for p in "universidad nacional de colombia".split()])
print(iniciales)   # UNDC
```

## 1.23 f-strings: cadenas formateadas

Un f-string (*formatted string literal*, literal de cadena formateada) es un string que lleva la letra `f` (o `F`) justo antes de las comillas y que puede contener expresiones de Python entre llaves `{}`. Cuando el programa se ejecuta, Python evalúa cada expresión, convierte su resultado en texto y lo inserta en ese lugar del string. A esta operación de insertar valores dentro de un texto se le llama interpolación.

```python
nombre = "Ana"
edad = 20
print(f"Me llamo {nombre} y tengo {edad} años")  # Me llamo Ana y tengo 20 años
```

Paso a paso:

1. La `f` antes de las comillas le indica a Python que el string contiene campos para reemplazar.
2. Python encuentra `{nombre}`, evalúa la variable y obtiene `"Ana"`.
3. Encuentra `{edad}`, obtiene el número `20` y lo convierte en el texto `"20"` automáticamente, sin necesidad de `str()`.
4. El resultado es un string normal: se puede imprimir, guardar en una variable o escribir en un archivo.

> Sin la `f`, las llaves se imprimen tal cual: `"Me llamo {nombre}"` muestra literalmente `Me llamo {nombre}`. Olvidar la `f` es el error más común con f-strings.

##### Origen de los f-strings

Combinar texto con valores es una de las tareas más frecuentes de cualquier programa, y Python ha tenido varias formas de hacerlo a lo largo de su historia:

| Forma | Desde | Origen | Ejemplo |
|-------|-------|--------|---------|
| Operador `%` | Primeras versiones (años 90) | Heredado de la función `printf` del lenguaje C | `"Hola %s, tienes %d años" % (nombre, edad)` |
| `string.Template` | Python 2.4 (2004) | PEP 292 | `Template("Hola $nombre").substitute(nombre=nombre)` |
| Método `format()` | Python 2.6 y 3.0 (2008) | PEP 3101 | `"Hola {}, tienes {} años".format(nombre, edad)` |
| f-strings | Python 3.6 (2016) | PEP 498 | `f"Hola {nombre}, tienes {edad} años"` |

Una PEP (*Python Enhancement Proposal*, propuesta de mejora de Python) es el documento con el que se propone y discute formalmente un cambio al lenguaje. Los f-strings se propusieron en la PEP 498, *Literal String Interpolation*, escrita por Eric V. Smith en 2015, y llegaron con Python 3.6, publicado en diciembre de 2016.

El problema que buscaban resolver era que las formas anteriores separan el texto de los valores: con `%` o con `format()`, el texto tiene marcadores vacíos (`%s` o `{}`) y los valores se escriben al final, así que hay que contar posiciones para saber qué valor va en cada marcador. Con muchos valores esto se vuelve difícil de leer y fácil de equivocar. Los f-strings ponen cada expresión exactamente donde va a aparecer, por lo que el código se lee casi igual que el resultado. Además, son la forma más rápida de las cuatro.

Los f-strings han seguido mejorando en versiones posteriores:

- Python 3.8 (2019) agregó el especificador `=`, que muestra el nombre de la expresión junto con su valor y es muy útil para depurar.
- Python 3.12 (2023), con la PEP 701, eliminó varias restricciones: desde esta versión se pueden usar dentro de las llaves las mismas comillas que encierran el f-string y escribir expresiones en varias líneas.

Las tres formas siguientes producen exactamente el mismo resultado:

```python
nombre = "Ana"
nota = 4.567

print("Estudiante: %s, nota: %.1f" % (nombre, nota))       # Estudiante: Ana, nota: 4.6
print("Estudiante: {}, nota: {:.1f}".format(nombre, nota))  # Estudiante: Ana, nota: 4.6
print(f"Estudiante: {nombre}, nota: {nota:.1f}")            # Estudiante: Ana, nota: 4.6
```

> Hoy los f-strings son la forma recomendada. Las otras siguen funcionando y aparecen en mucho código existente, por eso conviene reconocerlas.

##### Expresiones dentro de las llaves

Entre las llaves se puede escribir cualquier expresión de Python, es decir, cualquier cosa que produzca un valor: una variable, una operación, una llamada a una función o a un método, un índice o una expresión condicional.

```python
a = 7
b = 3
print(f"{a} + {b} = {a + b}")        # 7 + 3 = 10
print(f"{a} / {b} = {a / b}")        # 7 / 3 = 2.3333333333333335

nombre = "ana gómez"
print(f"Nombre: {nombre.title()}")   # Nombre: Ana Gómez
print(f"Letras: {len(nombre)}")      # Letras: 9

notas = [4.5, 3.2, 4.8]
print(f"Primera nota: {notas[0]}, promedio: {sum(notas) / len(notas)}")
# Primera nota: 4.5, promedio: 4.166666666666667

nota = 2.8
print(f"Estado: {'Aprobó' if nota >= 3.0 else 'Reprobó'}")  # Estado: Reprobó
```

Para acceder a un diccionario dentro de un f-string, la clave se escribe con comillas distintas a las del f-string:

```python
estudiante = {"nombre": "Luis", "nota": 3.2}
print(f"{estudiante['nombre']} sacó {estudiante['nota']}")  # Luis sacó 3.2
```

> Antes de Python 3.12, usar las mismas comillas (`f"{estudiante["nombre"]}"`) producía un error de sintaxis. Usar comillas distintas funciona en todas las versiones.

##### El especificador de formato

Después de la expresión se pueden escribir dos puntos y un especificador de formato, que indica cómo debe verse el valor: cuántos decimales, qué ancho ocupa, hacia dónde se alinea, si lleva separador de miles, etc.

```python
f"{valor:especificador}"
```

El especificador tiene varias partes opcionales que, si se usan, deben ir en este orden:

```
{ valor : relleno alineación signo 0 ancho , .precisión tipo }
```

| Parte | Qué indica | Ejemplos |
|-------|------------|----------|
| `relleno` | Carácter con que se completa el espacio sobrante (por defecto, espacio). Solo se puede usar si se indica también la alineación | `*`, `-`, `.` |
| `alineación` | Hacia dónde se ubica el valor dentro del ancho | `<` izquierda, `>` derecha, `^` centro |
| `signo` | Cuándo mostrar el signo del número | `+` siempre, `-` solo negativos (por defecto) |
| `0` | Rellenar con ceros a la izquierda | `05` |
| `ancho` | Cantidad mínima de caracteres que ocupa el resultado | `10` |
| `,` o `_` | Separador de miles | `,` |
| `.precisión` | Decimales para números; caracteres máximos para textos | `.2` |
| `tipo` | Cómo se presenta el valor | `f` decimal, `d` entero, `%` porcentaje, `e` científica |

Paso a paso de `f"{precio:>12,.2f}"` con `precio = 1234567.891`:

1. `>` alinea a la derecha.
2. `12` hace que el resultado ocupe 12 caracteres.
3. `,` agrega el separador de miles.
4. `.2f` muestra el número en formato decimal con 2 decimales.
5. Resultado: `'1,234,567.89'`, que tiene exactamente 12 caracteres.

```python
precio = 1234567.891
print(f"[{precio:>12,.2f}]")  # [1,234,567.89]
print(f"[{precio:>15,.2f}]")  # [   1,234,567.89]
```

> Los corchetes del f-string se usan solo para ver los espacios. Si el valor es más largo que el ancho, no se recorta: el ancho es un mínimo, no un máximo.

##### Decimales

```python
pi = 3.14159265
print(f"{pi:.2f}")   # 3.14
print(f"{pi:.4f}")   # 3.1416  ← redondea, no corta
print(f"{pi:.0f}")   # 3
print(f"{2.5:.0f}")  # 2       ← ver la nota
print(f"{10:.2f}")   # 10.00   ← también sirve para enteros
```

> `.0f` redondea al par más cercano cuando el número está exactamente en la mitad, por eso `2.5` se muestra como `2` y `3.5` como `4`. Además, algunos decimales no se pueden guardar exactos en el computador: `2.675` se guarda como `2.67499999...` y `f"{2.675:.2f}"` muestra `2.67`. Para cálculos con dinero donde esto importe se usa el módulo `decimal`.

##### Ancho y alineación

```python
for producto, precio in [("Café", 3500), ("Arepa", 2000), ("Buñuelo", 1500)]:
    print(f"|{producto:<10}|{producto:^10}|{producto:>10}|{precio:>8}|")
# |Café      |   Café   |      Café|    3500|
# |Arepa     |  Arepa   |     Arepa|    2000|
# |Buñuelo   | Buñuelo  |   Buñuelo|    1500|
```

Por defecto, los textos se alinean a la izquierda y los números a la derecha:

```python
print(f"[{'Ana':8}]")  # [Ana     ]
print(f"[{42:8}]")     # [      42]
```

Con un carácter de relleno se pueden hacer líneas, títulos y precios con puntos:

```python
print(f"{'MENÚ':=^30}")
print(f"{'Bandeja paisa':.<22}{28000:.>8}")
print(f"{'Ajiaco':.<22}{24000:.>8}")
print(f"{'':=^30}")
# =============MENÚ=============
# Bandeja paisa............28000
# Ajiaco...................24000
# ==============================
```

##### Separador de miles y porcentajes

```python
poblacion = 2612958
print(f"{poblacion:,}")      # 2,612,958
print(f"{poblacion:_}")      # 2_612_958
print(f"${1234.5:,.2f}")     # $1,234.50

aprobados = 0.8571
print(f"{aprobados:%}")      # 85.710000%
print(f"{aprobados:.1%}")    # 85.7%  ← multiplica por 100 y agrega el signo %
```

En Colombia el separador de miles es el punto y el decimal es la coma. Una forma sencilla de obtener ese formato es intercambiar los caracteres después de formatear:

```python
valor = 1234567.5
texto = f"{valor:,.2f}"   # '1,234,567.50'
colombiano = texto.replace(",", "X").replace(".", ",").replace("X", ".")
print(f"${colombiano}")   # $1.234.567,50
```

> Se usa un carácter temporal (`"X"`) porque, si se reemplazaran las comas por puntos directamente, ya no se podrían distinguir de los puntos decimales.

##### Signo y ceros a la izquierda

```python
for t in [5, -3, 0]:
    print(f"Variación: {t:+}")
# Variación: +5
# Variación: -3
# Variación: +0

print(f"{7:03}")             # 007
print(f"Factura N.° {42:06}")  # Factura N.° 000042
print(f"{-5:05}")            # -0005  ← el signo cuenta dentro del ancho
print(f"{9:02}:{5:02}")      # 09:05  ← útil para horas
```

##### Otros tipos de presentación

```python
n = 255
print(f"{n:b}")         # 11111111  ← binario
print(f"{n:o}")         # 377       ← octal
print(f"{n:x}")         # ff        ← hexadecimal en minúscula
print(f"{n:X}")         # FF        ← hexadecimal en mayúscula
print(f"{n:08b}")       # 11111111

distancia = 149600000000
print(f"{distancia:e}")    # 1.496000e+11
print(f"{distancia:.2e}")  # 1.50e+11  ← notación científica con 2 decimales
```

La precisión también funciona con textos: indica la cantidad máxima de caracteres que se muestran:

```python
titulo = "Fundamentos de Programación"
print(f"{titulo:.11}")      # Fundamentos
print(f"[{titulo:.11}...]")  # [Fundamentos...]
```

##### Fechas

Los objetos de fecha del módulo `datetime` aceptan sus propios códigos de formato dentro de un f-string. Cada código empieza con `%`:

```python
from datetime import date

hoy = date(2026, 10, 18)
print(f"{hoy}")             # 2026-10-18
print(f"{hoy:%d/%m/%Y}")    # 18/10/2026
print(f"{hoy:%d-%m-%y}")    # 18-10-26
print(f"Año: {hoy:%Y}")     # Año: 2026
```

| Código | Significado | Ejemplo |
|--------|-------------|---------|
| `%d` | Día con dos dígitos | `18` |
| `%m` | Mes con dos dígitos | `10` |
| `%Y` | Año con cuatro dígitos | `2026` |
| `%y` | Año con dos dígitos | `26` |
| `%H` / `%M` / `%S` | Hora (24 h), minutos y segundos | `14` / `32` / `05` |

> Los nombres de días y meses (códigos `%A` y `%B`) dependen del idioma configurado en el computador y por defecto aparecen en inglés.

##### Ancho y precisión variables

El ancho o la precisión también pueden venir de una variable, escribiéndola entre llaves dentro del especificador:

```python
ancho = 12
decimales = 3
valor = 3.14159
print(f"[{valor:>{ancho}.{decimales}f}]")  # [       3.142]
```

Esto permite ajustar una tabla al dato más largo:

```python
nombres = ["Ana", "Valentina", "Luis"]
ancho = max([len(n) for n in nombres])
for n in nombres:
    print(f"|{n:<{ancho}}|")
# |Ana      |
# |Valentina|
# |Luis     |
```

##### Depurar con `=`

Desde Python 3.8, escribir `=` después de la expresión muestra la expresión, el signo igual y su valor. Es muy útil para revisar qué valor tiene una variable mientras se busca un error:

```python
total = 43000
cantidad = 20
print(f"{total=}")                   # total=43000
print(f"{total / cantidad=}")        # total / cantidad=2150.0
print(f"{total / cantidad = :.1f}")  # total / cantidad = 2150.0
```

##### Representación con `!r`

Escribir `!r` después de la expresión muestra el valor como lo haría `repr()`. Con los strings, esto incluye las comillas y hace visibles los espacios y los caracteres especiales:

```python
dato = "  Ana\n"
print(f"Valor: {dato}.")    # muestra el salto de línea real
# Valor:   Ana
# .
print(f"Valor: {dato!r}.")  # Valor: '  Ana\n'.
```

##### Escribir llaves literales

Como las llaves marcan dónde hay una expresión, para escribir una llave como texto hay que duplicarla:

```python
n = 5
print(f"{{n}} vale {n}")       # {n} vale 5
print(f"Conjunto: {{{n}}}")    # Conjunto: {5}
```

##### f-strings de varias líneas

Con comillas triples, un f-string puede ocupar varias líneas. Es útil para reportes:

```python
nombre = "Valentina"
notas = [4.0, 3.5, 4.8]
promedio = sum(notas) / len(notas)

reporte = f"""
REPORTE DE NOTAS
Estudiante: {nombre}
Notas:      {', '.join([str(n) for n in notas])}
Promedio:   {promedio:.2f}
Estado:     {'Aprobó' if promedio >= 3 else 'Reprobó'}
"""
print(reporte)
#
# REPORTE DE NOTAS
# Estudiante: Valentina
# Notas:      4.0, 3.5, 4.8
# Promedio:   4.10
# Estado:     Aprobó
#
```

##### ⚠️ Errores comunes con f-strings

Olvidar la `f`. No produce ningún error, solo un resultado inesperado:

```python
nombre = "Ana"
print("Hola {nombre}")   # Hola {nombre}
print(f"Hola {nombre}")  # Hola Ana
```

Usar un tipo de formato que no corresponde al valor. El tipo `f` es para números y `d` es solo para enteros:

```python
print(f"{'Ana':.2f}")
# ValueError: Unknown format code 'f' for object of type 'str'
```

```python
print(f"{3.5:d}")
# ValueError: Unknown format code 'd' for object of type 'float'
```

Esto pasa con frecuencia con los datos leídos con `input()` o de un archivo, que siempre son strings. Hay que convertirlos primero:

```python
dato = "4.567"             # como si viniera de input()
print(f"{float(dato):.2f}")  # 4.57
```

Poner espacios en el especificador. Los espacios cambian el significado o producen error. Se escribe `{precio:>10,.2f}`, todo junto.

# 2\. Biblioteca `string`

Una biblioteca o módulo es un archivo con código ya escrito (funciones, constantes y otras herramientas) que se puede reutilizar en cualquier programa, sin tener que escribirlo de nuevo. Python trae instalada una colección de módulos llamada biblioteca estándar, con herramientas para matemáticas (`math`), números aleatorios (`random`), fechas (`datetime`), manejo de archivos (`os`) y muchas tareas más.

Para usar un módulo hay que importarlo con la instrucción `import`, normalmente al inicio del programa. Después de importarlo, sus elementos se usan escribiendo el nombre del módulo, un punto y el nombre del elemento, por ejemplo `string.digits`.

El módulo `string` reúne herramientas para trabajar con texto. Lo más usado de este módulo son sus constantes: variables que ya contienen strings con grupos de caracteres de uso frecuente, como el alfabeto, los dígitos o los signos de puntuación. Se llaman constantes porque su valor no está pensado para cambiar. Usarlas tiene dos ventajas: evita escribir esos grupos a mano (con el riesgo de olvidar algún carácter) y hace el código más fácil de leer, porque `c in string.digits` dice claramente qué se está revisando.

```python
import string
```

> No confundas el módulo `string` con el tipo `str`. Los métodos como `upper()` o `split()` pertenecen a `str` y no necesitan importar nada.


## 2.1 Constantes del módulo

```python
import string

print(string.ascii_lowercase)  # abcdefghijklmnopqrstuvwxyz
print(string.ascii_uppercase)  # ABCDEFGHIJKLMNOPQRSTUVWXYZ
print(string.ascii_letters)    # abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ
print(string.digits)           # 0123456789
print(string.hexdigits)        # 0123456789abcdefABCDEF
print(string.octdigits)        # 01234567
print(string.punctuation)      # !"#$%&'()*+,-./:;<=>?@[\]^_`{|}~
print(repr(string.whitespace)) # ' \t\n\r\x0b\x0c'
```

| Constante | Contiene |
|-----------|----------|
| `ascii_lowercase` | Las 26 letras minúsculas del alfabeto inglés |
| `ascii_uppercase` | Las 26 letras mayúsculas |
| `ascii_letters` | Las dos anteriores juntas |
| `digits` | Los dígitos del 0 al 9 |
| `hexdigits` | Los dígitos hexadecimales |
| `octdigits` | Los dígitos octales |
| `punctuation` | Los signos de puntuación del teclado inglés |
| `whitespace` | Los caracteres de espacio en blanco (espacio, `\t`, `\n`, etc.) |

Las constantes son strings normales, así que se usan con `in`, `len()`, índices y ciclos:

```python
import string

print(len(string.ascii_lowercase))     # 26
print(string.ascii_lowercase[0:5])     # abcde
print("?" in string.punctuation)       # True
print(string.ascii_uppercase.find("D"))  # 3  ← la posición de la letra en el alfabeto
```

⚠️ No incluyen la ñ, las vocales con tilde ni los signos de apertura del español:

import string

letras_es = string.ascii_lowercase + "ñáéíóúü"
puntuacion_es = string.punctuation + "¿¡«»"
print("ñ" in letras_es, "¿" in puntuacion_es)  # True True
print("ñ".isalpha())                            # True
```

## 2.3 Uso 1: clasificar caracteres

```python
import string

clave = "Medellin#2026"
mayus = minus = digitos = signos = 0
for c in clave:
    if c in string.ascii_uppercase:
        mayus += 1
    elif c in string.ascii_lowercase:
        minus += 1
    elif c in string.digits:
        digitos += 1
    elif c in string.punctuation:
        signos += 1
print(mayus, minus, digitos, signos)  # 1 7 4 1
```

> `mayus = minus = digitos = signos = 0` asigna el valor 0 a las cuatro variables en una sola línea.


## 2.4 Uso 2: quitar los signos de puntuación

Recorriendo el texto:

```python
import string

texto = "¡Hola, mundo! ¿Cómo están?"
limpio = ""
for c in texto:
    if c not in string.punctuation + "¡¿":
        limpio += c
print(limpio)  # Hola mundo Cómo están
```

Los strings también tienen el método `translate()`, que reemplaza o elimina varios caracteres a la vez en un solo paso. Para saber qué hacer con cada carácter, `translate()` usa una tabla de traducción: una estructura parecida a un diccionario que indica, para cada carácter, por cuál se debe reemplazar o si se debe eliminar. La tabla se crea con `str.maketrans()`:

```python
import string

texto = "¡Hola, mundo! ¿Cómo están?"

tabla = str.maketrans("", "", string.punctuation)
print(texto.translate(tabla))  # ¡Hola mundo ¿Cómo están  ← ¡ y ¿ no están en punctuation

tabla = str.maketrans("", "", string.punctuation + "¡¿")
print(texto.translate(tabla))  # Hola mundo Cómo están
```

`str.maketrans()` recibe tres argumentos:

1. Los caracteres que se van a reemplazar.
2. Los caracteres por los que se reemplazan, en el mismo orden (debe tener la misma longitud que el primero).
3. Los caracteres que se van a eliminar.

Con los dos primeros argumentos se pueden quitar las tildes:

```python
tildes = str.maketrans("áéíóúÁÉÍÓÚ", "aeiouAEIOU")
print("Canción Árbol ratón".translate(tildes))  # Cancion Arbol raton
```

## 2.5 Uso 3: recorrer el alfabeto

Con `ascii_lowercase` se obtiene la posición de una letra en el alfabeto y la letra que está en una posición dada:

import string

alfabeto = string.ascii_lowercase

palabra = "hola"

## 2.6 `string.capwords()`

La función `capwords()` del módulo `string` pone en mayúscula la primera letra de cada palabra y el resto en minúscula. Se diferencia del método `title()` en cómo decide dónde empieza una palabra: `title()` considera que una palabra empieza después de cualquier carácter que no sea letra (un apóstrofo, un guion, un número), mientras que `capwords()` solo considera palabras las partes separadas por espacios:

```python
import string

texto = "rock'n'roll en el parque"
print(texto.title())            # Rock'N'Roll En El Parque  ← trata ' como separador
print(string.capwords(texto))   # Rock'n'roll En El Parque
```

`capwords()` también deja un solo espacio entre palabras:

```python
import string
print(string.capwords("   ana    maría   gómez "))  # Ana María Gómez
```

# 3\. Archivos de texto

Las variables de un programa se guardan en la memoria RAM, que es muy rápida pero volátil: todo su contenido se pierde cuando el programa termina o el computador se apaga. Si un programa registra las ventas del día en una lista, esa lista desaparece al cerrarlo. Para que la información persista, es decir, para conservarla y poder recuperarla después, hay que guardarla en un almacenamiento permanente, como el disco duro o una memoria USB, en forma de archivo.

Un archivo es un conjunto de datos guardado en un dispositivo de almacenamiento e identificado con un nombre. El nombre suele terminar en una extensión (`.txt`, `.csv`, `.pdf`, `.jpg`) que indica qué tipo de contenido tiene. Los archivos se organizan en carpetas (también llamadas directorios), y la ubicación exacta de un archivo dentro de esas carpetas se llama su ruta.

Hay dos grandes tipos de archivos:

- Un archivo de texto contiene únicamente caracteres legibles, organizados en líneas. Se puede abrir y entender con cualquier editor de texto, como el Bloc de notas. Son archivos de texto los `.txt`, los `.csv`, los `.html` e incluso los programas de Python (`.py`). Desde Python, el contenido de un archivo de texto se maneja como un string, en el que el final de cada línea está marcado con el carácter `\n`.
- Un archivo binario contiene datos codificados que solo tienen sentido para el programa que los creó, como una imagen, un audio o un documento de Word. Al abrirlo con un editor de texto se ven símbolos sin sentido.

En esta sección se trabaja solo con archivos de texto. Trabajar con un archivo siempre tiene tres pasos:

1. Abrir el archivo con `open()`. Abrir es pedirle al sistema operativo acceso al archivo, indicando qué se quiere hacer con él (leer, escribir o agregar). El resultado es un objeto archivo: la "conexión" a través de la cual el programa lee o escribe.
2. Leer o escribir usando los métodos de ese objeto archivo.
3. Cerrar el archivo con `close()`, para terminar de guardar los cambios y liberar el acceso.


## 3.1 La instrucción `with`

La instrucción `with` sirve para trabajar con un recurso que hay que liberar cuando se termina de usar. Un recurso es algo que el programa toma prestado del sistema operativo mientras lo necesita: un archivo, una conexión a internet o a una base de datos. Si el programa no lo devuelve, ese recurso puede quedar bloqueado, o los datos pueden quedar incompletos.

`with` crea un bloque de contexto: un bloque de código indentado durante el cual el recurso está disponible. La instrucción se encarga de dos cosas automáticamente:

- Al entrar al bloque, prepara el recurso. En el caso de un archivo, lo abre.
- Al salir del bloque, lo libera. En el caso de un archivo, lo cierra.

Lo más importante es que la salida ocurre siempre: cuando el bloque termina normalmente y también cuando ocurre un error en medio del bloque. Así se garantiza que el archivo se cierra sin que el programador tenga que acordarse de hacerlo.

Una comparación útil es el préstamo de un libro en una biblioteca con devolución automática: al entrar al bloque se recibe el libro, dentro del bloque se lee, y al salir el libro se devuelve solo, aunque la lectura se haya interrumpido.

La estructura general para archivos es:

```python
with open(nombre_del_archivo, modo, encoding="utf-8") as variable:
    # instrucciones que usan el archivo a través de la variable
# aquí el archivo ya está cerrado
```

Cada parte cumple una función:

| Parte | Qué hace |
|-------|----------|
| `with` | Inicia el bloque de contexto |
| `open(...)` | Abre el archivo y devuelve un objeto archivo, la "conexión" a través de la cual se lee o se escribe |
| `as variable` | Le da un nombre al objeto archivo para usarlo dentro del bloque. Se suele llamar `archivo` o `f` (de *file*) |
| `:` | Indica que empieza el bloque, igual que en un `if` o un `for` |
| Bloque indentado | Las instrucciones que leen o escriben el archivo |
| Fin de la indentación | Python cierra el archivo automáticamente |

Un primer ejemplo completo, que escribe un archivo y luego lo lee:

```python
with open("saludo.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Hola, mundo\n")
    archivo.write("Esta es la segunda línea\n")
    archivo.write("Fin")

with open("saludo.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(contenido)
# Hola, mundo
# Esta es la segunda línea
# Fin
```

Paso a paso:

1. El primer `with` abre `saludo.txt` en modo escritura (`"w"`). Si el archivo no existe, lo crea.
2. Dentro del bloque, cada `write()` escribe una parte del texto.
3. Al terminar la indentación, `with` cierra el archivo y el texto queda guardado en el disco.
4. El segundo `with` abre el mismo archivo en modo lectura (`"r"`), y `read()` guarda todo su contenido en la variable `contenido`.
5. Al terminar ese bloque el archivo se cierra, pero la variable `contenido` sigue existiendo, porque es un string normal guardado en la memoria.

El objeto archivo tiene un atributo `closed` que indica si el archivo está cerrado. Fuera del bloque vale `True`, y el objeto ya no se puede usar para leer ni escribir:

```python
with open("saludo.txt", "r", encoding="utf-8") as archivo:
    print(archivo.closed)  # False ← dentro del bloque, el archivo está abierto

print(archivo.closed)      # True  ← al salir del bloque, se cerró solo
archivo.read()
# ValueError: I/O operation on closed file.
```

> Todo el trabajo con el archivo debe quedar dentro del bloque indentado. Lo que se lea del archivo se guarda en variables, y esas variables sí se pueden usar después.


## 3.2 ¿Por qué es tan importante cerrar un archivo?

Cerrar un archivo significa terminar la conexión entre el programa y el archivo. Al cerrarlo, Python termina de escribir en el disco cualquier dato que estuviera pendiente y le avisa al sistema operativo que el archivo ya está libre. Hay tres razones por las que este paso es indispensable, y por las que `with` es la forma recomendada de trabajar con archivos:

1. Para que los datos se guarden. Escribir en el disco es lento comparado con trabajar en la memoria, así que, por eficiencia, `write()` no escribe en el disco de inmediato: acumula el texto en un buffer, una zona de la memoria donde los datos esperan temporalmente, y lo guarda en el disco por bloques. Al cerrar el archivo se guarda lo que haya quedado pendiente en el buffer. Si el programa termina abruptamente sin cerrar el archivo, se pueden perder datos.
2. Para liberar el archivo. Mientras está abierto, otros programas pueden no poder usarlo; por ejemplo, en Windows no se puede borrar ni mover.
3. Porque hay un límite de archivos que un programa puede tener abiertos al mismo tiempo.

> En otros materiales puedes encontrar archivos abiertos con `open()` y cerrados manualmente con el método `close()`. `with` hace ese cierre por nosotros, y lo hace incluso si ocurre un error, por eso en este documento siempre se usa `with`.


## 3.3 La función `open()` y los modos de apertura

```python
with open("nombre_del_archivo.txt", "modo", encoding="utf-8") as f:
```

La función `open()` recibe tres datos:

- El nombre del archivo, que puede incluir la ruta de la carpeta (sección 3.11). Si no la incluye, el archivo se busca en la carpeta donde se está ejecutando el programa.
- El modo de apertura, que indica la intención con la que se abre el archivo: leer, escribir desde cero o agregar al final. El modo determina qué operaciones están permitidas y qué pasa con el contenido que el archivo ya tenía.
- La codificación, `encoding="utf-8"`, que indica cómo se guardan los caracteres especiales como la ñ y las tildes (sección 3.12).

| Modo | Nombre | Si el archivo existe | Si no existe | Permite |
|:----:|--------|----------------------|--------------|---------|
| `"r"` | Leer (*read*) | Lo abre para leer desde el inicio | `FileNotFoundError` | Leer |
| `"w"` | Escribir (*write*) | Borra todo su contenido | Lo crea | Escribir |
| `"a"` | Agregar (*append*) | Escribe al final, sin borrar nada | Lo crea | Escribir |
| `"x"` | Crear (*exclusive*) | `FileExistsError` | Lo crea | Escribir |

> Si no se indica el modo, Python usa `"r"`.


## 3.4 Escribir: `write()`

Escribir en un archivo significa enviar texto desde el programa hacia el archivo para que quede guardado en el disco. Para escribir, el archivo se debe abrir en modo `"w"`, `"a"` o `"x"`. El método `write(texto)` escribe el string que recibe en la posición donde esté el cursor del archivo: al inicio si se abrió con `"w"`, o al final si se abrió con `"a"`.

`write()` escribe el texto exactamente como se le da y no agrega el salto de línea: hay que escribir `\n` donde se quiera terminar una línea. En el ejemplo de la sección 3.1, el archivo `saludo.txt` quedó así:

```
Hola, mundo
Esta es la segunda línea
Fin
```

Si se olvidan los `\n`, todo queda en la misma línea:

```python
with open("pegado.txt", "w", encoding="utf-8") as f:
    f.write("uno")
    f.write("dos")
    f.write("tres")

with open("pegado.txt", "r", encoding="utf-8") as f:
    print(f.read())  # unodostres
```

`write()` solo recibe strings. Los números hay que convertirlos:

```python
with open("numeros.txt", "w", encoding="utf-8") as f:
    f.write(25)
# TypeError: write() argument must be str, not int
```

```python
with open("numeros.txt", "w", encoding="utf-8") as f:
    for n in range(1, 6):
        f.write(str(n) + "\n")                 # con str()
    f.write(f"Total: {1 + 2 + 3 + 4 + 5}\n")   # o con un f-string

with open("numeros.txt", "r", encoding="utf-8") as f:
    print(f.read(), end="")
# 1
# 2
# 3
# 4
# 5
# Total: 15
```

> `print(..., end="")` evita que `print()` agregue un salto de línea extra, porque el contenido del archivo ya termina en `\n`. `write()` devuelve la cantidad de caracteres que escribió; en un programa normal ese valor se ignora, pero en un notebook de Jupyter o Colab puede aparecer como un número debajo de la celda. No es un error.

`writelines()` escribe todos los strings de una lista, uno tras otro. Tampoco agrega los saltos de línea:

```python
frutas = ["mango\n", "guayaba\n", "lulo\n"]
with open("frutas.txt", "w", encoding="utf-8") as f:
    f.writelines(frutas)

with open("frutas.txt", "r", encoding="utf-8") as f:
    print(f.read(), end="")
# mango
# guayaba
# lulo
```

También se puede usar `print()` con el argumento `file`. A diferencia de `write()`, `print()` sí agrega el salto de línea y convierte los números automáticamente:

```python
with open("tabla_7.txt", "w", encoding="utf-8") as f:
    for i in range(1, 4):
        print(7, "x", i, "=", 7 * i, file=f)

with open("tabla_7.txt", "r", encoding="utf-8") as f:
    print(f.read(), end="")
# 7 x 1 = 7
# 7 x 2 = 14
# 7 x 3 = 21
```




## 3.5 Leer un archivo completo: `read()`

Leer un archivo significa traer su contenido desde el disco hacia el programa, para guardarlo en una variable y procesarlo. Para leer, el archivo se debe abrir en modo `"r"`. Python ofrece varias formas de leer, según si se necesita todo el contenido de una vez o una línea a la vez. La más sencilla es el método `read()`, que lee el archivo completo desde la posición del cursor hasta el final y lo devuelve como un solo string, incluidos los `\n`:

```python
with open("saludo.txt", "r", encoding="utf-8") as f:
    contenido = f.read()

print(repr(contenido))  # 'Hola, mundo\nEsta es la segunda línea\nFin'
print(len(contenido))   # 40
```

> `repr()` muestra el string con sus caracteres especiales visibles, por eso se ven los `\n`.

Como el contenido es un string, se le pueden aplicar todos los métodos de la sección 1:

```python
with open("saludo.txt", "r", encoding="utf-8") as f:
    contenido = f.read()

print(contenido.count("a"))     # 5
print(len(contenido.split()))   # 8  ← cantidad de palabras
print(contenido.upper())
# HOLA, MUNDO
# ESTA ES LA SEGUNDA LÍNEA
# FIN
```

## 3.6 Leer línea por línea

Muchas veces el contenido de un archivo tiene sentido línea por línea: cada línea es un registro, un contacto o una venta. Para estos casos hay tres opciones:

- `readline()` lee una sola línea cada vez que se llama, incluido el `\n` del final. Cuando ya no quedan líneas, devuelve un string vacío.
- `readlines()` lee todas las líneas y las devuelve en una lista, donde cada elemento es una línea con su `\n`.
- Un ciclo `for` sobre el objeto archivo entrega una línea en cada vuelta.

```python
with open("saludo.txt", "r", encoding="utf-8") as f:
    print(repr(f.readline()))  # 'Hola, mundo\n'
    print(repr(f.readline()))  # 'Esta es la segunda línea\n'
    print(repr(f.readline()))  # 'Fin'
    print(repr(f.readline()))  # ''  ← string vacío: se llegó al final

with open("saludo.txt", "r", encoding="utf-8") as f:
    print(f.readlines())  # ['Hola, mundo\n', 'Esta es la segunda línea\n', 'Fin']
```

La forma más común y eficiente es recorrer el archivo con un ciclo `for` dentro del bloque `with`:

```python
with open("saludo.txt", "r", encoding="utf-8") as f:
    for linea in f:
        print(linea)
# Hola, mundo
#
# Esta es la segunda línea
#
# Fin
```

Aparecen líneas en blanco de más porque cada línea ya trae su `\n` y `print()` agrega otro. Se corrige quitando el `\n` con `rstrip("\n")` (o con `strip()`, que además quita los espacios):

```python
with open("saludo.txt", "r", encoding="utf-8") as f:
    numero = 1
    for linea in f:
        linea = linea.rstrip("\n")
        print(f"{numero}: {linea}")
        numero += 1
# 1: Hola, mundo
# 2: Esta es la segunda línea
# 3: Fin
```

> Si el archivo es muy grande, el ciclo `for` es preferible a `read()` y `readlines()`, porque solo tiene una línea en memoria a la vez.


## 3.7 El cursor del archivo

Un archivo abierto tiene un cursor (también llamado puntero de posición), que funciona como el cursor de un editor de texto: marca el punto exacto del archivo donde ocurrirá la próxima lectura o escritura. Al abrir un archivo en modo `"r"` o `"w"`, el cursor está al inicio; en modo `"a"`, está al final.

Cada operación de lectura avanza el cursor hasta donde terminó de leer, de modo que la siguiente lectura continúa desde ahí. Cuando el cursor llega al final del archivo, cualquier lectura adicional devuelve un string vacío. Por eso un segundo `read()` dentro del mismo bloque ya no encuentra nada:

```python
with open("saludo.txt", "r", encoding="utf-8") as f:
    primera = f.readline()
    resto = f.read()          # continúa donde quedó readline()
    otra_vez = f.read()       # el cursor ya está al final

print(repr(primera))   # 'Hola, mundo\n'
print(repr(resto))     # 'Esta es la segunda línea\nFin'
print(repr(otra_vez))  # ''
```

Para volver a leer desde el inicio se abre el archivo en un nuevo bloque `with`, o se usa `f.seek(0)` dentro del mismo bloque, que mueve el cursor al inicio:

```python
with open("saludo.txt", "r", encoding="utf-8") as f:
    primera_vez = f.read()
    f.seek(0)                 # el cursor vuelve al inicio
    segunda_vez = f.read()

print(primera_vez == segunda_vez)  # True
```

## 3.8 Agregar al final: modo `"a"`

Agregar (*append*) significa escribir al final de un archivo conservando todo lo que ya contenía. El modo `"w"` borra el contenido cada vez que se abre el archivo; el modo `"a"`, en cambio, ubica el cursor al final, de modo que lo nuevo se escribe después del contenido existente. Si el archivo no existe, se crea vacío. Es el modo adecuado para archivos que crecen con el tiempo, como una bitácora, un registro de ventas o un historial de eventos:

```python
with open("bitacora.txt", "w", encoding="utf-8") as f:
    f.write("Lunes: clase de listas\n")

with open("bitacora.txt", "a", encoding="utf-8") as f:
    f.write("Martes: clase de strings\n")

with open("bitacora.txt", "a", encoding="utf-8") as f:
    f.write("Miércoles: clase de archivos\n")

with open("bitacora.txt", "r", encoding="utf-8") as f:
    print(f.read(), end="")
# Lunes: clase de listas
# Martes: clase de strings
# Miércoles: clase de archivos
```

## 3.9 El modo `"w"` borra el archivo al abrirlo

⚠️ El contenido se borra en el momento de abrir el archivo, aunque no se escriba nada dentro del bloque:

```python
with open("bitacora.txt", "w", encoding="utf-8") as f:
    pass   # pass no hace nada: el bloque queda vacío

with open("bitacora.txt", "r", encoding="utf-8") as f:
    print(repr(f.read()))  # ''  ← se perdió todo
```

> Antes de abrir un archivo con `"w"`, asegúrate de que no contiene información que necesites. Si quieres evitar sobrescribir un archivo por accidente, usa el modo `"x"`.

## 3.11 Rutas de archivos

La ruta de un archivo es la dirección que indica dónde está guardado. Igual que la dirección de una casa, nombra en orden las carpetas que hay que recorrer para llegar a él, separadas por `/` (o por `\` en Windows). Hay dos tipos de rutas:

- Una ruta relativa, como `"datos.txt"` o `"carpeta/datos.txt"`, se interpreta a partir de la carpeta desde donde se ejecuta el programa, llamada directorio de trabajo. Es la forma más común, porque el programa sigue funcionando aunque toda la carpeta del proyecto se mueva a otro lugar. La función `os.getcwd()` del módulo `os` muestra cuál es el directorio de trabajo actual.
- Una ruta absoluta indica la ubicación completa desde la raíz del disco, como `"C:/Users/ana/Documentos/datos.txt"` en Windows o `"/home/ana/datos.txt"` en Linux y macOS.

En Windows, las rutas se escriben con `\`, que en un string es el inicio de una secuencia de escape (sección 1.3). Por ejemplo, en `"C:\nuevos\tareas.txt"` Python interpreta `\n` como un salto de línea y `\t` como una tabulación. Hay tres soluciones:

```python
ruta1 = "C:/nuevos/tareas.txt"      # usar / (Python lo acepta también en Windows)
ruta2 = "C:\\nuevos\\tareas.txt"    # escribir \\ para cada barra
ruta3 = r"C:\nuevos\tareas.txt"     # string "crudo": la r desactiva las secuencias de escape
print(ruta3)  # C:\nuevos\tareas.txt
```

> En Google Colab, los archivos se guardan en la carpeta `/content` del servidor y se borran al terminar la sesión. Para conservarlos hay que descargarlos o conectar Google Drive.

## 3.12 ¿Por qué `encoding="utf-8"`?

El disco no guarda letras, sino números (bytes). La codificación (*encoding*) es la regla que establece cómo se traduce cada carácter a bytes al escribir un archivo, y cómo se traducen esos bytes de vuelta a caracteres al leerlo. Las letras del inglés y los dígitos se codifican igual en casi todas las codificaciones, pero la ñ, las vocales con tilde y otros caracteres especiales no.

UTF-8 es la codificación más usada en el mundo: puede representar todos los caracteres de Unicode y es el estándar en la web, en Linux y en macOS. Si un archivo se escribe con una codificación y se lee con otra, los caracteres especiales se dañan:

```python
with open("ciudad.txt", "w", encoding="utf-8") as f:
    f.write("Medellín, Antioquia. Año 2026")

with open("ciudad.txt", "r", encoding="latin-1") as f:   # codificación incorrecta
    print(f.read())  # MedellÃ­n, Antioquia. AÃ±o 2026
```

Si no se indica la codificación, Python usa la predeterminada del sistema operativo, que en Windows suele ser distinta de UTF-8. Por eso conviene escribir siempre `encoding="utf-8"`, tanto al escribir como al leer.


## 3.13 Ejemplo completo: guardar y leer una tabla

Un formato muy usado para guardar tablas en archivos de texto es CSV (*Comma-Separated Values*, valores separados por comas). En un archivo CSV, cada línea es una fila de la tabla y los valores de cada columna se separan con comas. La primera línea suele contener los encabezados, es decir, los nombres de las columnas.

Como es un archivo de texto plano, un CSV se puede abrir con cualquier editor de texto, pero también con Excel o Google Sheets, que lo muestran como una hoja de cálculo. Por eso es un formato muy usado para intercambiar datos entre programas. Se crea y se lee con lo visto hasta ahora: `join()` para escribir cada fila y `split()` para separarla al leer.

Escribir la tabla de estudiantes de la unidad de tablas:

```python
estudiantes = [["Ana", 20, 4.5],
               ["Luis", 22, 3.2],
               ["Sofía", 19, 4.8]]

with open("estudiantes.csv", "w", encoding="utf-8") as f:
    f.write("nombre,edad,nota\n")   # encabezados
    for fila in estudiantes:
        f.write(",".join([str(dato) for dato in fila]) + "\n")
```

El archivo `estudiantes.csv` queda así:

```
nombre,edad,nota
Ana,20,4.5
Luis,22,3.2
Sofía,19,4.8
```

Leerla y reconstruir la tabla:

```python
tabla = []
with open("estudiantes.csv", "r", encoding="utf-8") as f:
    encabezados = f.readline().strip().split(",")   # la primera línea
    for linea in f:                                  # el resto de las líneas
        nombre, edad, nota = linea.strip().split(",")
        tabla.append([nombre, int(edad), float(nota)])

print(encabezados)  # ['nombre', 'edad', 'nota']
print(tabla)        # [['Ana', 20, 4.5], ['Luis', 22, 3.2], ['Sofía', 19, 4.8]]

promedio = sum([fila[2] for fila in tabla]) / len(tabla)
print(f"Promedio: {promedio:.2f}")  # Promedio: 4.17
```

Paso a paso de la lectura de la línea `"Luis,22,3.2\n"`:

1. `strip()` quita el `\n` del final: `"Luis,22,3.2"`.
2. `split(",")` la divide: `["Luis", "22", "3.2"]`.
3. Se desempaqueta en tres variables, que todavía son strings.
4. `int()` y `float()` convierten la edad y la nota para poder hacer cálculos.

> Todo lo que se lee de un archivo de texto es un string, igual que lo que se lee con `input()`. Los números hay que convertirlos.


## 3.14 Ejemplo completo: procesar un archivo y escribir otro

Un patrón muy común es leer un archivo, transformar su contenido y guardar el resultado en otro archivo. Un solo `with` puede abrir varios archivos a la vez, separándolos con comas; todos se cierran al salir del bloque:

```python
with open("ventas.txt", "w", encoding="utf-8") as f:
    f.write("café 3500 4\n")
    f.write("arepa 2000 10\n")
    f.write("buñuelo 1500 6\n")

total = 0
with open("ventas.txt", "r", encoding="utf-8") as entrada, \
     open("reporte.txt", "w", encoding="utf-8") as salida:
    salida.write("REPORTE DE VENTAS\n")
    for linea in entrada:
        producto, precio, cantidad = linea.split()
        subtotal = int(precio) * int(cantidad)
        total += subtotal
        salida.write(f"{producto.capitalize():<10}{subtotal:>8,}\n")
    salida.write(f"{'TOTAL':<10}{total:>8,}\n")

with open("reporte.txt", "r", encoding="utf-8") as f:
    print(f.read(), end="")
# REPORTE DE VENTAS
# Café        14,000
# Arepa       20,000
# Buñuelo      9,000
# TOTAL       43,000
```

> La `\` al final de la línea del `with` permite continuar la instrucción en la línea siguiente.


## 3.15 Resumen de archivos

| Operación | Código |
|-----------|--------|
| Abrir para leer | `with open("a.txt", "r", encoding="utf-8") as f:` |
| Abrir para escribir (borra el contenido) | `with open("a.txt", "w", encoding="utf-8") as f:` |
| Abrir para agregar al final | `with open("a.txt", "a", encoding="utf-8") as f:` |
| Abrir varios archivos a la vez | `with open("a.txt") as f1, open("b.txt", "w") as f2:` |
| Escribir un texto | `f.write("texto\n")` |
| Escribir una lista de textos | `f.writelines(lista)` |
| Escribir con `print()` | `print(valor, file=f)` |
| Leer todo como un string | `f.read()` |
| Leer una línea | `f.readline()` |
| Leer todas las líneas en una lista | `f.readlines()` |
| Recorrer las líneas | `for linea in f:` |
| Volver al inicio del archivo | `f.seek(0)` |
| Saber si el archivo está cerrado | `f.closed` |
| Cerrar el archivo | Automático al terminar el bloque `with` |
