# 1\. Listas anidadas



## 1.1 Definición y creación de una lista anidada

Una lista anidada es una lista cuyos elementos son, a su vez, otras listas. Se usa para representar información con varias dimensiones, como una cuadrícula, una matriz o un tablero.

```python
matriz = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]
```

Paso a paso:

1. `matriz` es una lista con 3 elementos.
2. Cada elemento es una lista con 3 números. A cada una de estas sublistas la llamamos fila.
3. La posición de un número dentro de su fila es su columna.

|          | columna 0 | columna 1 | columna 2 |
|----------|:---------:|:---------:|:---------:|
| fila 0 | 1 | 2 | 3 |
| fila 1 | 4 | 5 | 6 |
| fila 2 | 7 | 8 | 9 |

## 1.2 Acceder a los elementos

Con un índice se obtiene una fila completa. Con dos índices, `matriz[fila][columna]`, se obtiene un elemento:

```python
matriz = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]

print(matriz[1])      # [4, 5, 6]  ← la fila 1 completa
print(matriz[1][2])   # 6          ← fila 1, columna 2
print(matriz[0][0])   # 1          ← primera fila, primera columna
print(matriz[-1][-1]) # 9          ← última fila, última columna
```

Python evalúa `matriz[1][2]` de izquierda a derecha:

1. `matriz[1]` devuelve la lista `[4, 5, 6]`.
2. `[4, 5, 6][2]` devuelve `6`.

## 1.3 Tamaño: filas y columnas

```python
matriz = [[1, 2, 3],
          [4, 5, 6]]

filas = len(matriz)        # cantidad de sublistas
columnas = len(matriz[0])  # cantidad de elementos de la primera fila
print(filas, columnas)     # 2 3
```

> `len(matriz)` **no** cuenta todos los números: cuenta cuántas filas hay. Para contar todos los elementos hay que sumar el tamaño de cada fila.

## 1.4 Modificar elementos

Para modificar elementos usamos el acceso indexado de `lista[fila][columna]`

```python
matriz = [[1, 2, 3],
          [4, 5, 6]]
matriz[0][1] = 20
print(matriz)  # [[1, 20, 3], [4, 5, 6]]

## 1.5 Agregar filas y elementos

```python
matriz = [[1, 2],
          [3, 4]]

matriz.append([5, 6])   # agrega una fila nueva al final
print(matriz)           # [[1, 2], [3, 4], [5, 6]]

matriz[0].append(99)    # agrega un elemento al final de la fila 0
print(matriz)           # [[1, 2, 99], [3, 4], [5, 6]]
```

Nota: usar append sobre filas puede desbalancear el estado cuadrado de una matriz.

## 1.6 Recorrer una lista anidada

Se usan dos ciclos `for`: el exterior recorre las filas y el interior recorre los elementos de cada fila.

```python
matriz = [[1, 2, 3],
          [4, 5, 6]]

for fila in matriz:
    for valor in fila:
        print(valor, end=" ")
    print()  # salto de línea al terminar cada fila
# 1 2 3
# 4 5 6
```

Paso a paso:

1. En la primera vuelta del ciclo exterior, `fila` vale `[1, 2, 3]`.
2. El ciclo interior imprime `1`, `2` y `3` en la misma línea, porque `end=" "` reemplaza el salto de línea por un espacio.
3. `print()` hace el salto de línea.
4. El ciclo exterior pasa a `[4, 5, 6]` y se repite el proceso.

Si se necesitan las posiciones, se recorren los índices:

```python
matriz = [[1, 2, 3],
          [4, 5, 6]]

for i in range(len(matriz)):
    for j in range(len(matriz[i])):
        print(f"matriz[{i}][{j}] = {matriz[i][j]}")
# matriz[0][0] = 1
# matriz[0][1] = 2
# matriz[0][2] = 3
# matriz[1][0] = 4
# matriz[1][1] = 5
# matriz[1][2] = 6
```

## 1.7 Aplanar una lista anidada

Aplanar es convertir la lista anidada en una lista simple con todos los elementos:

```python
matriz = [[1, 2, 3],
          [4, 5, 6]]

plana = []
for fila in matriz:
    for valor in fila:
        plana.append(valor)
print(plana)  # [1, 2, 3, 4, 5, 6]
```

## 1.8 Copiar una lista anidada

El método `copy()` hace una copia superficial: crea una lista exterior nueva, pero las filas siguen siendo las mismas.

```python
a = [[1, 2], [3, 4]]
b = a.copy()
b[0][0] = 99
print(a)  # [[99, 2], [3, 4]]  ← también cambió la original
```

Para copiar también las filas se usa `deepcopy()` del módulo `copy`:

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][0] = 99
print(a)  # [[1, 2], [3, 4]]  ← la original no cambia
print(b)  # [[99, 2], [3, 4]]
```

# 2\. Tablas


## 2.1 Definición de Tabla

Una tabla es una lista anidada en la que:

- Cada fila es un registro (por ejemplo, un estudiante).
- Cada columna es un campo o dato de ese registro (nombre, edad, nota).

A diferencia de una matriz numérica, las columnas de una tabla pueden tener tipos de datos distintos. Los nombres de las columnas se suelen guardar en una lista aparte llamada encabezados.

```python
encabezados = ["Nombre", "Edad", "Nota"]
estudiantes = [["Ana", 20, 4.5],
               ["Luis", 22, 3.2],
               ["Sofía", 19, 4.8],
               ["Carlos", 21, 2.9]]
```

````python
encabezados = ["Nombre", "Edad", "Nota"]
estudiantes = [["Ana", 20, 4.5],
               ["Luis", 22, 3.2],
               ["Sofía", 19, 4.8],
               ["Carlos", 21, 2.9]]
````

## 2.2 Acceder a registros y campos

```python
print(estudiantes[2])     # ['Sofía', 19, 4.8]  ← registro completo
print(estudiantes[2][0])  # Sofía               ← nombre de ese registro
print(estudiantes[1][2])  # 3.2                 ← nota de Luis
```

Para no tener que recordar qué número corresponde a cada columna, se pueden usar constantes:

```python
NOMBRE, EDAD, NOTA = 0, 1, 2
print(estudiantes[1][NOTA])  # 3.2
```

## 2.3 Desempaquetar las filas en el ciclo

Si todas las filas tienen la misma cantidad de columnas, se pueden asignar a varias variables directamente en el `for`:

```python
for nombre, edad, nota in estudiantes:
    print(nombre, "tiene", edad, "años y nota", nota)
# Ana tiene 20 años y nota 4.5
# Luis tiene 22 años y nota 3.2
# Sofía tiene 19 años y nota 4.8
# Carlos tiene 21 años y nota 2.9
```

## 2.4 Extraer una columna

Para obtener todos los valores de una columna se toma el mismo índice en cada fila:

```python
notas = []
for fila in estudiantes:
    notas.append(fila[2])
print(notas)  # [4.5, 3.2, 4.8, 2.9]
```

## 2.5 Cálculos sobre una columna
Se conserva el concepto del numeral anterior y usamos un acumulador.

```python
suma = 0
for fila in estudiantes:
    suma += fila[2]
promedio = suma / len(estudiantes)
print(f"Promedio del grupo: {promedio:.2f}")  # Promedio del grupo: 3.85
```

## 2.6 Agregar una columna calculada

Se recorre cada fila y se le agrega un valor al final. También se actualizan los encabezados:

```python
encabezados.append("Estado")
for fila in estudiantes:
    if fila[2] >= 3.0:
        fila.append("Aprobó")
    else:
        fila.append("Reprobó")

print(encabezados)     # ['Nombre', 'Edad', 'Nota', 'Estado']
print(estudiantes[3])  # ['Carlos', 21, 2.9, 'Reprobó']
```

## 2.7 Buscar registros

Es importante recordar que la búsqueda con coincidencia es exacta, con mayúsculas y minúsculas.

```python
buscado = "Sofía"
for fila in estudiantes:
    if fila[0] == buscado:
        print("Encontrado:", fila)
        break
else:
    print("No existe un estudiante con ese nombre")
# Encontrado: ['Sofía', 19, 4.8, 'Aprobó']
```

## 2.8 Ordenar la tabla por una columna

`sort()` compara listas elemento por elemento, así que por defecto ordena por la primera columna. Para ordenar por otra columna se usa el argumento `key` con una función que indique qué valor comparar. Una forma corta de escribir esa función es con `lambda`:

```python
estudiantes.sort(key=lambda fila: fila[2], reverse=True)
for fila in estudiantes:
    print(fila)
# ['Sofía', 19, 4.8, 'Aprobó']
# ['Ana', 20, 4.5, 'Aprobó']
# ['Luis', 22, 3.2, 'Aprobó']
# ['Carlos', 21, 2.9, 'Reprobó']
```

`lambda fila: fila[2]` es una función sin nombre que recibe una fila y devuelve su nota. Es equivalente a:

```python
def obtener_nota(fila):
    return fila[2]

estudiantes.sort(key=obtener_nota, reverse=True)
```

El concepto de función lo veremos en clases posteriores, de momento identificamos el comportamiento por su estructura y lo aplicamos.

## 2.9 Tablas numéricas: totales por fila y por columna

En una tabla de ventas, las filas pueden ser productos y las columnas, meses:

```python
productos = ["Café", "Arepas", "Buñuelos"]
meses = ["Ene", "Feb", "Mar"]
ventas = [[120, 150, 90],
          [200, 180, 210],
          [80, 95, 110]]
```

Total por fila (cada producto): se suman los elementos de cada sublista.

```python
for i in range(len(ventas)):
    total_fila = 0
    for valor in ventas[i]:
        total_fila += valor
    print(productos[i], total_fila)
# Café 360
# Arepas 590
# Buñuelos 285
```

Total por columna (cada mes): el ciclo **exterior** recorre las columnas y el **interior** recorre las filas.

```python
for j in range(len(meses)):
    total_columna = 0
    for i in range(len(ventas)):
        total_columna += ventas[i][j]
    print(meses[j], total_columna)
# Ene 400
# Feb 425
# Mar 410
```

Todo junto, con formato de tabla:

```python
print(f"{'Producto':<10}", end="")
for mes in meses:
    print(f"{mes:>6}", end="")
print(f"{'Total':>7}")

for i in range(len(ventas)):
    print(f"{productos[i]:<10}", end="")
    for valor in ventas[i]:
        print(f"{valor:>6}", end="")
    print(f"{sum(ventas[i]):>7}")
# Producto     Ene   Feb   Mar  Total
# Café         120   150    90    360
# Arepas       200   180   210    590
# Buñuelos      80    95   110    285
```

# 3\. List Comprehension


## 3.1 Definición de List Comprehension

Una list comprehension (lista por comprensión) es una forma compacta de crear una lista nueva a partir de otra secuencia, escrita en una sola línea.

```python
nueva_lista = [expresion for elemento in iterable]
```

## 3.2 De un ciclo `for` a una list comprehension

Estos dos códigos hacen exactamente lo mismo:

```python
# Con un ciclo for
cuadrados = []
for n in range(1, 6):
    cuadrados.append(n ** 2)
print(cuadrados)  # [1, 4, 9, 16, 25]
```

```python
# Con list comprehension
cuadrados = [n ** 2 for n in range(1, 6)]
print(cuadrados)  # [1, 4, 9, 16, 25]
```

Paso a paso para convertir el ciclo:

1. Se abre la lista con `[`. Ya no hace falta crear `cuadrados = []` ni llamar a `append()`.
2. Se escribe primero lo que se agregaría con `append()`: `n ** 2`.
3. Se escribe el encabezado del ciclo, sin los dos puntos: `for n in range(1, 6)`.
4. Se cierra con `]`.

## 3.3 Aplicar una operación a cada elemento

```python
precios = [10000, 30000, 50000]
con_iva = [p * 1.19 for p in precios]
print(con_iva)  # [11900.0, 35700.0, 59500.0]

heroes = ["batman", "flash", "superman"]
mayusculas = [h.upper() for h in heroes]
print(mayusculas)  # ['BATMAN', 'FLASH', 'SUPERMAN']
```

## 3.4 Filtrar con `if`

Se agrega una condición al final. Solo se incluyen los elementos que la cumplen:

```python
nueva_lista = [expresion for elemento in iterable if condicion]
```

```python
numeros = [5, 1, 6, 8, 11, 3, 1, 99]
pares = [n for n in numeros if n % 2 == 0]
print(pares)  # [6, 8]
```

Equivale a:

```python
pares = []
for n in numeros:
    if n % 2 == 0:
        pares.append(n)
```

## 3.5 Elegir entre dos valores con `if ... else`

Si se quiere transformar cada elemento según una condición (y no descartar ninguno), el `if ... else` va antes del `for`:

```python
nueva_lista = [valor_si if condicion else valor_no for elemento in iterable]
```

```python
numeros = [5, 1, 6, 8]
tipos = ["par" if n % 2 == 0 else "impar" for n in numeros]
print(tipos)  # ['impar', 'impar', 'par', 'par']
```

> `if` al final → filtra: la lista resultante puede tener menos elementos.
> `if ... else` al inicio → transforma: la lista resultante tiene la misma cantidad de elementos.
> Escribir un `else` al final, como `[n for n in numeros if n > 3 else 0]`, produce un `SyntaxError`.

## 3.6 List comprehension con listas anidadas

Aplanar una matriz. Los `for` se escriben en el mismo orden que en los ciclos anidados tradicionales:

```python
matriz = [[1, 2, 3],
          [4, 5, 6]]

plana = [valor for fila in matriz for valor in fila]
print(plana)  # [1, 2, 3, 4, 5, 6]
```

Equivale a:

```python
plana = []
for fila in matriz:           # primer for
    for valor in fila:        # segundo for
        plana.append(valor)
```

Crear una matriz. Aquí hay una list comprehension dentro de otra. Esta es la forma correcta de evitar el error de `[[0] * 3] * 3` visto en la sección 2.22:

```python
ceros = [[0 for j in range(3)] for i in range(2)]
print(ceros)  # [[0, 0, 0], [0, 0, 0]]

ceros[0][0] = 1
print(ceros)  # [[1, 0, 0], [0, 0, 0]]  ← cada fila es independiente
```

La comprensión exterior crea las filas. La interior se evalúa de nuevo en cada fila y crea una lista nueva cada vez.

Tabla de multiplicar usando las variables de ambos ciclos:

```python
tabla = [[i * j for j in range(1, 4)] for i in range(1, 4)]
print(tabla)  # [[1, 2, 3], [2, 4, 6], [3, 6, 9]]
```

Transformar cada elemento de una matriz conservando su forma:

```python
matriz = [[1, 2, 3],
          [4, 5, 6]]
dobles = [[valor * 2 for valor in fila] for fila in matriz]
print(dobles)  # [[2, 4, 6], [8, 10, 12]]
```

> Observa la diferencia: `[x for fila in m for x in fila]` (dos `for` seguidos) produce una lista plana. `[[x for x in fila] for fila in m]` (una comprensión dentro de otra) produce una lista anidada.

## 3.7 Cuándo no usar list comprehension

- Cuando no se va a crear una lista. Si solo se quiere imprimir o ejecutar una acción, se usa un `for` normal. `[print(x) for x in lista]` funciona, pero crea una lista de `None` sin ningún propósito.
- Cuando la lógica es compleja. Si hay varias condiciones, varios pasos o más de dos niveles de anidamiento, un ciclo `for` tradicional es más fácil de leer y de depurar.

```python
# Difícil de leer
r = [x * y for x in range(5) if x % 2 == 0 for y in range(5) if y > x if x + y < 6]

# Más claro con ciclos
r = []
for x in range(5):
    if x % 2 == 0:
        for y in range(5):
            if y > x and x + y < 6:
                r.append(x * y)
```
