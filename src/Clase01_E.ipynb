### 1.¿Qué es programar y por qué es importante?



#### 1.1 ¿Qué es programar?

**Programar** es el proceso de escribir instrucciones precisas y ordenadas para que una computadora realice una tarea específica. Estas instrucciones se escriben en un **lenguaje de programación** (como Python, C++, Java, etc.), que es un lenguaje intermedio entre el pensamiento humano y el lenguaje que la máquina realmente entiende (código binario: ceros y unos).

Piensa en programar como escribir una **receta de cocina extremadamente detallada**: no basta con decir "hacer arroz", hay que especificar cada paso: cuánta agua, cuánto tiempo, a qué temperatura, en qué orden. Las computadoras no "intuyen" nada —> hacen exactamente lo que se les indica, ni más ni menos. Por eso programar exige ser **claro, ordenado y lógico**.

En términos más formales, programar consiste en:

1. **Analizar un problema** y entender qué se necesita resolver.
2. **Diseñar una solución** (un algoritmo: una secuencia de pasos lógicos).
3. **Traducir esa solución** a un lenguaje que la máquina pueda ejecutar (el código).
4. **Probar y corregir errores** (a esto se le llama *debugging*).

Primer caso de un [bug](https://americanhistory.si.edu/collections/object/nmah_334663) (error) en un programa. Clic [aquí](https://americanhistory.si.edu/collections/object/nmah_334663)

#### 1.3 Teorema del Programa Estructurado

##### ¿Qué dice el teorema?

El **teorema del programa estructurado** (formulado por Böhm y Jacopini en 1966) establece que:

> **Cualquier algoritmo o programa, sin importar qué tan complejo sea, puede construirse usando únicamente tres estructuras de control básicas:**
> 1. **Secuencia**
> 2. **Selección** (decisión)
> 3. **Iteración** (repetición)

Con solo estas tres estructuras, combinadas y anidadas de la forma correcta, puedes construir cualquier programa, sin importar su complejidad.

---

## Las tres estructuras básicas

### 1. Secuencia
Es la ejecución de instrucciones **una tras otra**, en el orden en que aparecen escritas, de arriba hacia abajo.

```
Instrucción 1
Instrucción 2
Instrucción 3
```

📌 *Ejemplo:* leer dos números, sumarlos, mostrar el resultado.

### 2. Selección (decisión)
Permite que el programa **elija entre diferentes caminos** según se cumpla o no una condición. Es la base de la lógica condicional.

```
Si (condición) entonces
    Instrucción A
Sino
    Instrucción B
Fin Si
```

📌 *Ejemplo:* si la nota es mayor o igual a 3.0, el estudiante aprueba; si no, reprueba.

### 3. Iteración (repetición o bucle)
Permite **repetir un bloque de instrucciones** mientras se cumpla (o hasta que se cumpla) una condición determinada.

```
Mientras (condición) hacer
    Instrucción
Fin Mientras
```

📌 *Ejemplo:* sumar los números del 1 al 100 usando un bucle, en vez de escribir 100 líneas de código.

---

## ¿Por qué es importante este teorema?

**1. Es la base de la programación estructurada.**
Gracias a este teorema, surgió un paradigma de programación (la *programación estructurada*) que reemplazó el uso caótico del `GOTO`, muy común en los años 60-70, por un flujo de control claro, predecible y organizado.

**2. Mejora la legibilidad del código.**
Un programa construido con estas tres estructuras es mucho más fácil de leer, entender y seguir mentalmente, porque el flujo va de arriba hacia abajo sin "saltos" impredecibles a otras partes del código.

**3. Facilita la depuración (debugging) y el mantenimiento.**
Al tener un flujo lógico y ordenado, es más sencillo encontrar errores y modificar el programa sin romper otras partes del código.

**4. Es la base de todo lenguaje de programación moderno.**
Python, C++, Java, JavaScript, etc. todos estos lenguajes están diseñados alrededor de estas tres estructuras. Cuando usas un `if`, un `while`, un `for`, o simplemente escribes líneas de código en orden, estás aplicando este teorema sin darte cuenta.

**5. Permite demostrar formalmente que cualquier problema computable tiene solución algorítmica** usando solo estas herramientas es decir, le da una base matemática sólida a la disciplina de la programación.

---

## Tabla resumen

| Estructura | Palabra clave típica | ¿Qué hace? |
|---|---|---|
| Secuencia | (ninguna, es el orden natural) | Ejecuta instrucciones una tras otra |
| Selección | `if`, `else`, `switch` | Elige un camino según una condición |
| Iteración | `while`, `for`, `do-while` | Repite instrucciones mientras se cumpla una condición |

### 2.¿Por qué Python?

#### 2.1 ¿Qué es el ranking TIOBE?

El **[TIOBE](https://www.tiobe.com/tiobe-index/) Index** (o Índice TIOBE) es uno de los rankings más reconocidos a nivel mundial para medir la **popularidad de los lenguajes de programación**. Es publicado mensualmente por la empresa TIOBE Software (Países Bajos) desde el año 2001.

**¿Cómo se calcula?** No mide cuál lenguaje tiene "más código escrito" ni cuál es "el mejor" técnicamente. En cambio, analiza cuántas veces se busca cada lenguaje en más de 25 motores de búsqueda (Google, Bing, Wikipedia, YouTube, Amazon, entre otros). La lógica es simple: **mientras más se busca un lenguaje (tutoriales, cursos, documentación, solución de errores), más se está usando y enseñando en el mundo real.**

> 💡 Es decir, [TIOBE](https://www.tiobe.com/tiobe-index/) mide **relevancia e interés global**, no calidad técnica.

### 3.Esquema general de un programa (Entrada -> Procesamiento -> Salida)


Casi todo programa de computadora, sin importar qué tan simple o complejo sea, sigue un **patrón básico de tres fases**. Entrada->Procesamiento->Salida.

> **La idea central es sencilla:** un programa recibe datos, hace algo con ellos, y entrega un resultado.

Piensa en una **licuadora**: le metes ingredientes (entrada), la licuadora los procesa (procesamiento), y obtienes un jugo (salida). Todo programa de software funciona bajo esa misma lógica.

---

#### Las tres fases explicadas

##### 1. Entrada (Input)

Es la fase donde el programa **recibe los datos** que va a necesitar para trabajar. Estos datos pueden venir de distintas fuentes:

- Lo que el usuario escribe por teclado
- Un archivo (por ejemplo, un `.txt`)
- Un sensor (en sistemas embebidos o de control)
- Una base de datos
- Otro programa o una Aplicación


##### 2. Procesamiento (Process)

Es el "cerebro" del programa. Aquí es donde se **transforman los datos de entrada** usando operaciones, cálculos, comparaciones, algoritmos, estructuras de control (secuencia, selección, iteración), etc.

Esta es la fase donde realmente ocurre la "magia": se aplican fórmulas, se organiza información, se toman decisiones, se repiten cálculos, etc.


##### 3. Salida (Output)

Es la fase final, donde el programa **muestra o entrega el resultado** obtenido tras el procesamiento. Al igual que la entrada, la salida puede tomar distintas formas:

- Texto mostrado en pantalla
- Un archivo generado
- Una gráfica o visualización
- Una señal enviada a un dispositivo (por ejemplo, encender un motor)
- Datos enviados a otro sistema

---

| Fase | Qué hace en este ejemplo |
|---|---|
| **Entrada** | Captura el nombre y la edad ingresados por el usuario |
| **Procesamiento** | Evalúa si la edad es mayor o igual a 18 y construye el mensaje correspondiente |
| **Salida** | Muestra en pantalla el mensaje final |

---

##### ¿Por qué es importante este esquema para ti como estudiante?

**1. Te ayuda a planear antes de programar.**
Antes de escribir una sola línea de código, puedes preguntarte: *¿qué datos necesito recibir?, ¿qué debo hacer con ellos?, ¿qué debo mostrar al final?* Esto evita que te "pierdas" al programar.

**2. Es la base para diseñar algoritmos.**
Todo diagrama de flujo, pseudocódigo o algoritmo que construyas en tus primeras materias de programación estará organizado, casi siempre, siguiendo esta misma estructura.

**3. Aplica a cualquier tipo de programa, sin importar su tamaño.**
Desde un programa que suma dos números, hasta un sistema completo de inteligencia artificial, la lógica de fondo es la misma: **entran datos → se procesan → sale un resultado**.

**4. Facilita identificar errores.**
Cuando un programa no funciona como se espera, este esquema te ayuda a ubicar en qué fase está el problema: ¿estoy recibiendo mal los datos?, ¿el procesamiento tiene un error de lógica?, ¿o el problema es cómo muestro la salida?

---

> 💡 **Idea clave para recordar:** cada vez que vayas a resolver un problema con programación, hazte estas tres preguntas en orden:
> 1. **¿Qué datos necesito recibir?** (Entrada)
> 2. **¿Qué debo hacer con esos datos?** (Procesamiento)
> 3. **¿Qué resultado debo mostrar?** (Salida)
>
> Si logras responder estas tres preguntas con claridad, ya tienes el 80% del diseño de tu programa resuelto, solo falta traducirlo a código.

### 4.Algoritmo, Instrucciones y Código fuente. Lenguaje compilado vs Lenguaje interpretado.


#### 4.1 Algoritmo, Instrucciones y Código fuente

```
┌─────────────┐        ┌─────────────────┐        ┌────────────────┐
│  ALGORITMO  │  ───>  │  INSTRUCCIONES  │  ───>  │ CÓDIGO FUENTE  │
│  (el plan)  │        │ (pasos          │        │ (instrucciones │
│             │        │  ejecutables    │        │  completas en  │
│             │  ───>  │  individuales)  │  ───>  │  un código)    │
└─────────────┘        └─────────────────┘        └────────────────┘
```
---
| Concepto | ¿Qué es? | ¿Depende de un lenguaje? | Analogía |
|---|---|---|---|
| **Algoritmo** | El plan o la lógica para resolver un problema | No | La receta escrita en tu idioma |
| **Instrucción** | Un paso individual y ejecutable | Sí | Un paso específico de la receta ("bata 2 huevos") |
| **Código fuente** | El conjunto completo de instrucciones en un archivo | Sí | El recetario completo, listo para seguir |

---

* ¿Qué es un Algoritmo?
>Un algoritmo es un conjunto prescrito de instrucciones o reglas bien definidas, ordenadas y finitas que permite realizar una actividad mediante pasos sucesivos que no generen dudas a quien deba realizar dicha actividad.
Dados un estado inicial y una entrada, siguiendo los pasos sucesivos se llega a un estado final y se obtiene una solución.

##### Finalidad de los algoritmos

* Solución de problemas
> Los algoritmos se utilizan todos los días en cada una de nuestras actividades, cada problema de la vida diaria se puede representar de manera algorítmica.

##### Características de un buen algoritmo

>* Debe tener un punto particular de inicio.
>* Debe ser definido, no debe permitir dobles interpretaciones.
>* Debe ser general, es decir, soportar la mayoría de las variantes que se puedan presentar en la definición del problema.
>* Debe ser finito en tamaño y tiempo de ejecución.

En Resumen
*   Precisión: Los pasos a seguir en el algoritmo deben ser precisados claramente.
*   Determinismo: El algoritmo, dado un conjunto de datos idénticos de entrada siempre debe arrojar los mismos resultados.
*   Finitud: El algoritmo, independiente de la complejidad del mismo, siempre debe ser de longitud finita.

Nota: se puede ser tan específico o tan simple como se desee.


---

#### 5.6 Convención de nombres en programación

Es un conjunto de reglas para la elección de la secuencia de caracteres que se utiliza para identificadores que denoten variables, tipos, funciones y otras entidades en el código fuente y la documentación.

Algunas de las razones para utilizar una convención de nombres (en lugar de permitir a los programadores elegir cualquier secuencia de caracteres) son:
* Reducir el esfuerzo necesario para leer y entender el código fuente.
* Mejorar la apariencia del código fuente (por ejemplo, al no permitir nombres excesivamente largos o abreviaturas poco claros).

La elección de las convenciones de nombres puede ser un problema de enorme polémica, donde los partidarios de cada convención consideran la suya como la mejor y las demás inferiores. Coloquialmente, se dice que es una cuestión de dogma. Muchas empresas también han establecido su propio conjunto de convenciones para satisfacer mejor sus intereses.

### **En Python**
Para Python recomiendan [UpperCamelCase](https://es.wikipedia.org/wiki/Camel_case) para nombres de clases, CAPITALIZED_WITH_UNDERSCORES para las constantes y lowercase_separated_by_underscores para otros nombres. En Python, si un nombre está destinado a ser "privado", que está precedido de un guión bajo.

[Wikipedia](https://es.wikipedia.org/wiki/Convenci%C3%B3n_de_nombres_(programaci%C3%B3n))

##### Notacion Húngara
En programación informática, la notación húngara es un sistema usado normalmente para crear los nombres de variables. También se utiliza para nombrar las instancias de objetos en lenguajes de programación visuales, como por ejemplo Delphi. El nombre de la notación proviene del hecho de que su inventor, Charles Simonyi, nació en Hungría.
Esta convención es muy poco utilizada en las viejas versiones de Delphi pero es muy utilizada por los programadores de Microsoft y, en particular, en la programación del sistema operativo Windows.
Consiste en prefijos en minúsculas que se añaden a los nombres de las variables y que indican su tipo. El resto del nombre indica, lo más claramente posible, la función que realiza la variable.

[Wikipedia](https://es.wikipedia.org/wiki/Notaci%C3%B3n_h%C3%BAngara)

##### Recomendaciones

1.   Variables:
>*   Utilizar nombres descriptivos y en minúsculas.
>*   Para nombres compuestos, separar las palabras con guiones bajos (_). Ejemplo: mi_variable.

2.   Constantes:
>*   Escribir todas las letras en mayúsculas y separar palabras con guiones bajos.
>*   Ejemplo: MI_CONSTANTE.

3.   Funciones:
>*   Usar nombres descriptivos y en minúsculas.
>*   Para nombres compuestos, utilizar guiones bajos para separar palabras. Ejemplo: mi_funcion().
>*   Los nombres deben reflejar la acción que realiza la función.

4.   Clases:
>*   Usar el estilo CamelCase, donde la primera letra de cada palabra está en mayúsculas y no hay separación entre palabras. Ejemplo: MiClase.
>*   Los nombres de clases deben ser sustantivos o frases nominales.

5.   Módulos y paquetes:
>*   Nombres cortos, en minúsculas.
>*   Para nombres compuestos, es preferible no usar guiones bajos, aunque es aceptable.
>*   Ejemplo: mimodulo.

6.   Métodos de instancia y estáticos:
>*   Seguir las mismas convenciones que para las funciones.
>*   Los nombres de métodos privados deben comenzar con un guión bajo (_).
>*   Ejemplo: _mi_metodo_privado().

7.   Parámetros de función y método:
>*   Usar nombres descriptivos y en minúsculas.
>*   Si es posible, nombres cortos y significativos.

8.   Evitar:
>*   Usar caracteres únicos, excepto para contadores o iteradores.
>*   Usar nombres que puedan confundirse con nombres de variables internas de Python (como id, type, list).


Estas convenciones están en línea con la guía de estilo de Python, PEP 8, que es ampliamente aceptada y seguida en la comunidad de Python. Seguir estas prácticas ayuda a otros desarrolladores a entender tu código fácilmente.

### 6.Tipos de datos y conversiones

#### 6.1 Tipos de Datos

Los tipos de datos en Python son las diferentes categorías de valores que pueden ser almacenados y manipulados en Python. Cada tipo de dato define el rango de valores que puede tomar, las operaciones que se pueden realizar con ellos y la forma en que se almacenan.

Los tipos de datos más comunes en Python son:

*   Números:
>*   Enteros (**int**): Representan números enteros, sin parte fraccional. Ejemplo: 5, -3, 42.
>*   Números de punto flotante (**float**): Representan números reales con parte decimal. Ejemplo: 3.14, -0.001, 2.0.
>*   Números complejos (**complex**): Números que tienen una parte real y una imaginaria, útiles en cálculos matemáticos avanzados. Ejemplo: 2 + 3j.

*   Secuencias:
>*   Cadenas de texto (**str**): Secuencias de caracteres usadas para representar texto. Ejemplo: "Hola", "Python".
>*   Listas (**list**): Colecciones ordenadas y modificables de elementos, que pueden ser de distintos tipos. Ejemplo: [1, 2.5, 'Python'].
>*   Tuplas (**tuple**): Colecciones ordenadas e inmutables de elementos. Ejemplo: (1, 'a', 3.14).

*   Mapeados:
>*   Diccionarios (**dict**): Colecciones no ordenadas de pares clave-valor. Ejemplo: {'nombre': 'Juan', 'edad': 30}.

*   Conjuntos:
>*   Conjuntos (**set**): Colecciones no ordenadas de elementos únicos. Ejemplo: {1, 2, 3}.
>*   Conjuntos Inmutables (frozenset): Como los conjuntos, pero inmutables.

*   Booleanos (**bool**): Representan dos valores:
>*   Verdadero (True) y Falso (False).

*   NoneType:
>*   None (**NoneType**): Especialmente designado para denotar la ausencia de un valor.

Cada tipo de dato tiene sus propias características y métodos asociados, lo que permite realizar diferentes tipos de operaciones y manipulaciones en Python.

| Descripción | Generador (Clase) Python |
|---|---|
| Text Type: | `str` |
| Numeric Types: | `int`, `float`, `complex` |
| Sequence Types: | `list`, `tuple`, `range` |
| Mapping Type: | `dict` |
| Set Types: | `set`, `frozenset` |
| Boolean Type: | `bool` |
| None Type: | `NoneType` |

#### 6.2 Conversiones

##### Las funciones de conversión implícita y explícita

Si desea especificar el tipo de datos de una variable, puede hacerlo con conversión. Para realizar conversiones utilizamos funciones integradas de tipología de datos, de las cuales las más usadas son (sea `x` una variable a convertir):

*   [bool(x)](https://docs.python.org/3/library/functions.html#bool)
>*   Devuelve un valor booleano, es decir, uno de verdadero o falso. `x` se convierte utilizando el procedimiento estándar de prueba de verdad. Si `x` es falso o se omite, devuelve `False`; de lo contrario, devuelve `True`. La clase `bool` es una subclase de `int`. No se puede subclasificar más. Sus únicas instancias son `False` y `True`.

*   [float(x)](https://docs.python.org/3/library/functions.html#float)
>*   Devuelve un número de coma flotante construido a partir de un número o cadena `x`.
Si el argumento es una cadena, debe contener un número decimal, opcionalmente precedido por un signo y opcionalmente incrustado en espacios en blanco. El signo opcional puede ser `+` o `-` ; un signo `+` no tiene ningún efecto sobre el valor producido. El argumento también puede ser una cadena que represente un NaN (no un número) o un infinito positivo o negativo.

*   [int(x)](https://docs.python.org/3/library/functions.html#int)
>*   Devuelve un objeto entero construido a partir de un número o cadena `x`, o devuelve `0` si no se dan argumentos. Si `x` define `__int__()`, `int(x)` devuelve `x` `__int__()`. Si `x` define `__index__()`, devuelve `x` .`__index__()`. Si `x` define `__trunc__()`, devuelve `x` .`__trunc__()`. Para números de coma flotante, esto se trunca hacia cero.
Si `x` no es un número o si se proporciona una base, entonces `x` debe ser una instancia de cadena, `bytes` o `bytearray` que represente un literal entero en  [radix (base)](https://en.wikipedia.org/wiki/Radix). Opcionalmente, el literal puede ir precedido por `+` o `-` (sin espacios entre ellos) y rodeado por espacios en blanco. Un literal de base `n` consta de los dígitos del `0` al `n-1`, y de la a a la z (o de la A a la Z) tienen valores de 10 a 35. La base predeterminada es 10. Los valores permitidos son 0 y 2–36. Opcionalmente, los literales base-2, -8 y -16 pueden tener el prefijo 0b / 0B, 0o / 0O o 0x / 0X, como ocurre con los literales enteros en el código. Base 0 significa interpretar exactamente como un código literal, de modo que la base real es 2, 8, 10 o 16, y de modo que int ('010', 0) no es legal, mientras que int ('010') es, así como int ('010', 8).

*   [str(x)](https://docs.python.org/3/library/functions.html#str)
>*   Devuelve una versión de cadena de texto del objeto. Si no se proporciona el objeto, devuelve la cadena vacía. De lo contrario, el comportamiento de `str()` depende de si se proporciona codificación o errores, como sigue.
Si no se da ni codificación ni errores, `str(objeto)` devuelve `objeto.__str__()`, que es la representación de cadena "informal" o bien imprimible del objeto. Para los objetos de cadena, esta es la cadena en sí. Si el objeto no tiene un método `__str__()`, entonces `str()` vuelve a devolver repr (objeto).
